ИИ в обычном чате. Часть 2: управление многошаговым диалогом

В первой части чат-бот преобразовывал содержание: сокращал, переводил, объяснял, извлекал, генерировал и критиковал.

Теперь задача сложнее. Нужно не получить один ответ, а организовать последовательную работу:

```text
неясная проблема
→ уточнение
→ варианты
→ выбор направления
→ промежуточный результат
→ критика
→ исправление
→ проверяемое действие
```

Для всего материала достаточно обычного текстового чата без поиска, файлов, кода, агентов и других инструментов.

# Большую задачу не нужно решать одним ответом

Запрос:

```text
Сделай мне выступление о влиянии ИИ на образование.
```

Модель может сразу написать гладкий текст. Но до написания не были решены главные вопросы:

- для кого выступление;
- какую проблему оно рассматривает;
- какую позицию защищает;
- что должно измениться после выступления;
- на чём основан вывод;
- какие возражения нужно выдержать.

## Шаг 1. Сузить тему

```text
Мне нужно десятиминутное выступление
о влиянии ИИ на образование.

Не пиши выступление сразу.

Предложи пять узких тем,
между которыми есть реальный выбор позиции.

Для каждой укажи:
- главный вопрос;
- возможный тезис;
- целевую аудиторию;
- какие данные понадобятся;
- почему тема помещается в десять минут.
```

## Шаг 2. Уточнить тезис

Допустим, выбрана тема:

> ИИ помогает сильным студентам и маскирует пробелы у слабых.

```text
Помоги уточнить тезис.

Раздели:
- проверяемые утверждения;
- термины, которые нужно определить;
- предполагаемые причинные механизмы;
- альтернативные объяснения;
- ценностные суждения;
- более осторожную формулировку,
  которую легче обосновать.

Пока не пиши выступление.
```

## Шаг 3. Построить аргумент

```text
Построй карту аргумента без литературного текста:

главный тезис
→ основания
→ необходимые доказательства
→ ограничения
→ сильнейшее возражение
→ возможный ответ.

Не выдумывай исследования и статистику.
В местах, где нужен внешний источник,
поставь [НУЖЕН ИСТОЧНИК].
```

## Шаг 4. Атаковать конструкцию

```text
Найди только замечания,
способные изменить вывод или формулировку тезиса.

Проверь:
- самое слабое звено аргумента;
- причинные выводы без основания;
- скрытые ценностные предпочтения;
- контрпример;
- данные, которые могли бы опровергнуть тезис.

Не спорь ради количества замечаний.
```

## Шаг 5. Написать текст

```text
Теперь подготовь текст десятиминутного выступления.

Используй утверждённую карту аргумента.
Не добавляй новых фактов.

В начале поставь проблему,
затем тезис, основания и сильнейшее возражение.
В конце назови ограничение вывода
и практическое следствие.
```

Один большой запрос превратился в серию контролируемых решений. Пользователь видит, на каком шаге возникла ошибка.

# Когда задача ещё не сформулирована

Иногда человек открывает чат и знает только тему:

```text
Мне нужно сделать что-то про нейросети для семинара.
```

Вместо немедленной генерации можно начать с выяснения задачи.

```text
Я хочу подготовить что-то по теме нейросетей,
но пока не понимаю, какой результат мне нужен.

Не предлагай решение сразу.

Задавай по одному вопросу, чтобы выяснить:
- для кого результат;
- зачем он нужен;
- что должно измениться после его использования;
- сколько у меня времени;
- что я уже знаю;
- какой результат я способен проверить.

Не задавай больше семи вопросов.
После каждого моего ответа выбирай следующий вопрос.
```

## Найти более простую задачу

```text
Я думаю, что мне нужно [описание результата].

Но я могу путать желаемый формат
с реальной задачей.

Спроси:
- какую проблему я решаю;
- для кого;
- почему существующий способ не подходит;
- какой минимальный результат уже будет полезен.

После моих ответов предложи три возможные постановки задачи.
```

## Вопросы, которые действительно меняют результат

```text
Перед выполнением задачи задай не более трёх вопросов.

Выбирай только те вопросы,
разные ответы на которые существенно изменят результат.

Не спрашивай то, что можно безопасно указать как допущение.
```

# Few-shot: показать правило примерами

Иногда пример объясняет желаемый результат лучше длинной инструкции.

## Классификация обращений

```text
Размечай обращения по образцу.

ПРИМЕР 1
Обращение:
«Код подтверждения приходит через десять минут,
войти невозможно».

Результат:
Тема: авторизация
Симптом: задержка SMS-кода
Последствие: вход заблокирован
Серьёзность: высокая
Неизвестное: оператор, регион, время попытки

ПРИМЕР 2
Обращение:
«Добавьте тёмную тему».

Результат:
Тема: интерфейс
Симптом: отсутствует желаемая функция
Последствие: не указано
Серьёзность: пожелание
Неизвестное: причина запроса

Теперь обработай следующие обращения
в том же формате:
[вставить обращения]
```

Примеры задали:

- поля;
- степень подробности;
- способ различать симптом и причину;
- различие проблемы и пожелания;
- правило помечать неизвестное.

## Положительный и отрицательный пример

```text
ХОРОШИЙ ЗАГОЛОВОК
«Почему средний чек вырос,
хотя клиенты не стали покупать больше».

Почему хороший:
- содержит конкретное противоречие;
- обещает объяснение;
- не скрывает тему.

ПЛОХОЙ ЗАГОЛОВОК
«Вы не поверите, что случилось с нашими продажами!»

Почему плохой:
- скрывает содержание;
- обещает эмоцию вместо информации;
- не подходит деловой аудитории.

Создай десять заголовков к тексту ниже
по принципу хорошего примера.
Не копируй его синтаксис буквально.

Текст:
[вставить текст]
```

## Пограничные случаи

```text
Ниже правила классификации и примеры.

Если новый случай не подходит ни к одному классу,
не выбирай ближайший автоматически.
Напиши «нужна ручная проверка»
и объясни, каких сведений не хватает.
```

Few-shot повышает управляемость. Но если пример содержит ошибку, модель будет последовательно воспроизводить её.

# Управление длинным разговором

У модели ограничен доступный контекст. Его точный размер и способ обработки зависят от сервиса. Даже если весь диалог технически помещается, модель может путать версии, терять ограничения и продолжать отменённое решение.

Полезнее следить не за количеством сообщений, а за признаками проблем:

- возвращаются отброшенные идеи;
- путаются старая и новая версии;
- исчезают важные ограничения;
- разные ответы противоречат друг другу;
- задача фактически сменилась;
- пользователь сам уже не помнит состояние работы.

## Снимок состояния

```text
Остановись и составь текущее состояние нашей работы.

Цель:
Подтверждённые факты:
Принятые решения:
Действующие ограничения:
Отброшенные варианты:
Открытые вопросы:
Следующий шаг:
Непроверенные предположения:

Не продолжай решать задачу.
Сначала покажи состояние для моей проверки.
```

После проверки этот блок можно оставить в текущем чате или перенести в новый.

## Найти конфликтующий контекст

```text
Проведи аудит истории разговора.

Найди:
- противоречащие инструкции;
- отменённые решения;
- устаревшие версии текста;
- предположения, используемые как факты;
- требования, смысл которых изменился.

Не исправляй автоматически.
Сначала покажи конфликты.
```

## Когда открыть новый чат

Новый чат полезен, если:

- начинается другая задача;
- нужна независимая критика;
- накопилась спорная рамка;
- история мешает, а не помогает;
- есть проверенный снимок актуального состояния.

Новый чат удаляет влияние старой беседы, но не превращает модель в независимого эксперта.

# Несколько независимых веток вместо одного ответа

Одну идею можно отправить в несколько новых чатов.

## Чат 1: сильнейшая версия

```text
Ниже идея проекта.

Собери наиболее сильную реалистичную версию.
Можно изменить форму, но не скрывай стоимость и ограничения.

Идея:
[вставить идею]
```

## Чат 2: причины провала

```text
Ниже идея проекта.

Найди три наиболее вероятные причины провала.
Для каждой укажи ранний наблюдаемый сигнал.

Идея:
[вставить ту же идею]
```

## Чат 3: проверка

```text
Не оценивай идею в целом.

Предложи самый дешёвый эксперимент,
который способен опровергнуть её ключевое допущение.

Идея:
[вставить ту же идею]
```

## Чат 4: синтез

```text
Ниже три независимых разбора одной идеи.

Сопоставь их:
- где они согласны;
- какие предпосылки используют;
- где противоречат друг другу;
- что требует проверки;
- какое действие даст больше всего новой информации.

Не усредняй позиции автоматически.

[вставить три ответа]
```

Несколько чатов уменьшают влияние одного предыдущего ответа. Но это не несколько независимых людей: модели могут повторять одинаковые ошибки.

# Расширенный мозговой штурм

Задача:

> Небольшая кофейня рядом с университетом переполнена с 8:30 до 9:00, а после 11:00 почти пустует. Помещение расширить нельзя. Нужно увеличить дневную выручку, не ухудшив утреннюю очередь.

Хороший брейншторм не начинается со списка идей.

## Проход 1. Построить карту задачи

```text
Пока не предлагай решения.

Разбери задачу кофейни:

1. Какие разные причины могут объяснять
   отсутствие клиентов после 11:00?
2. Какие группы клиентов ведут себя по-разному?
3. Какие ограничения даны явно?
4. Какие ограничения мы только предполагаем?
5. Каких данных не хватает?
6. Какие рычаги доступны:
   продукт, цена, время, канал продажи,
   повторные покупки, партнёрства,
   обслуживание, свободные мощности?
```

## Проход 2. Генерировать по независимым направлениям

```text
Предложи решения по восьми направлениям:

1. перенести часть утреннего спроса;
2. привлечь новую аудиторию после 11:00;
3. продавать другие продукты;
4. продавать вне помещения;
5. использовать предзаказ;
6. создавать повторную покупку;
7. использовать партнёрства;
8. зарабатывать на свободной мощности кухни и персонала.

По три идеи на направление.

Каждая идея должна объяснять:
кто, за что, почему и когда заплатит.

Не оценивай идеи.
Не повторяй скидку под разными названиями.
```

## Проход 3. Перевернуть предпосылки

```text
Найди скрытые предпосылки в идеях.

Например:
- клиент обязательно приходит в кофейню;
- платят за одну чашку;
- продукт обязательно готовит бариста;
- продажа происходит перед употреблением;
- покупатель и потребитель — один человек.

Выбери пять сильных предпосылок.
Для каждой придумай решение,
в котором она не выполняется.
```

## Проход 4. X1 → X2 → X3

```text
Рассмотри задачу независимо в трёх ролях.

X1 — владелец кофейни:
ищет прибыль и простоту операции.

X2 — постоянный студент:
оценивает удобство, цену и реальное поведение клиентов.

X3 — операционный менеджер:
видит нагрузку персонала, запасы и риск хаоса.

Сначала каждый предлагает по пять идей,
не видя оценок остальных.

Затем каждый критикует идеи двух других
со своей позиции.

После перекрёстной критики
собери пять гибридных решений.

Не изображай роли как мнения реально опрошенных людей.
```

## Проход 5. Адвокат дьявола и адвокат сильной версии

```text
Для каждой идеи проведи два разбора.

АДВОКАТ ДЬЯВОЛА
- почему идея провалится;
- какую новую проблему создаст;
- какая скрытая стоимость возникнет;
- какое допущение наиболее сомнительно.

АДВОКАТ СИЛЬНОЙ ВЕРСИИ
- при каких условиях идея сработает;
- как исправить её сильнейшую слабость;
- с какой другой идеей её объединить;
- какая минимальная версия сохраняет ценность.

Не выбирай победителя.
```

## Проход 6. Удалить смысловые дубликаты

```text
Сгруппируй идеи по механизму,
а не по формулировке.

Покажи:
- общий механизм группы;
- почти одинаковые идеи;
- действительно отличающиеся варианты.

Оставь не более двенадцати содержательно разных идей.
```

## Проход 7. Перейти к экспериментам

```text
Выбери портфель из трёх экспериментов:

1. почти бесплатный и быстрый;
2. потенциально наиболее прибыльный;
3. проверяющий самое важное неизвестное о клиентах.

Для каждого укажи:
- гипотезу;
- действие;
- срок;
- метрику;
- критерий продолжения;
- критерий остановки.

Эксперименты не должны мешать друг другу.
```

Модель не узнала, чего хотят реальные посетители. Она помогла построить пространство гипотез и способы столкнуть их с реальностью.

# Актор-критик: создавать и проверять разными проходами

## Актор

```text
Подготовь черновик статьи
«Как удалённая работа влияет на молодых специалистов».

Не выбирай победителя заранее.
Отдели возможные преимущества,
риски и условия, при которых эффект меняется.
```

## Критик в новом чате

```text
Проведи содержательную критику текста.

Не переписывай его и не изображай суровый тон.

Найди:
- слабый аргумент;
- неподтверждённый факт;
- причинный вывод без основания;
- пропущенный сильный контраргумент;
- место, где позиция автора скрыта под видом факта;
- условие, при котором вывод перестаёт работать.

Текст:
[вставить черновик]
```

## Автор принимает не всё

```text
Ниже текст и замечания критика.

Для каждого замечания определи:
- подтверждается текстом;
- зависит от позиции критика;
- требует внешней проверки;
- несущественно для главного тезиса.

Предложи исправления только для существенных замечаний.
Не меняй позицию статьи ради формального баланса.
```

Разделение ролей помогает снизить якорение на первом варианте. Оно не делает критику автоматически правильной.

# Сценарное мышление вместо одного пророчества

Запрос «получится ли проект?» провоцирует один уверенный рассказ о будущем.

Полезнее исследовать разные ветви.

```text
Компания собирается использовать ИИ
для подготовки ответов клиентам.
Руководитель ожидает сократить нагрузку поддержки на 40%.

Не прогнозируй одно будущее.

Построй три сценария:
1. существенный успех;
2. частичный эффект;
3. провал.

Для каждого укажи:
- необходимые предпосылки;
- ранние признаки;
- возможные побочные эффекты;
- решение, которое потребуется принять;
- данные, которые нужно собирать с начала работы.
```

## Красная и синяя команды

```text
КРАСНАЯ КОМАНДА
Покажи, почему проект не достигнет результата.

СИНЯЯ КОМАНДА
Измени проект так,
чтобы он выдержал наиболее сильные возражения,
не скрывая стоимость и ограничения.

После двух раундов выдели:
- снятые возражения;
- замаскированные возражения;
- неизвестные, требующие проверки.
```

Сценарии не предсказывают будущее. Они помогают заранее увидеть условия, сигналы и возможные решения.

# Помощь в решении без передачи решения модели

Слабый запрос:

```text
Стоит ли мне соглашаться на это предложение?
```

Модель знает только описание пользователя и легко примет его рамку.

Более полезный процесс:

```text
Не принимай решение за меня.

Помоги построить сравнение вариантов.

1. Сначала выясни мои критерии.
2. Найди конфликты между критериями.
3. Попроси определить их относительную важность.
4. Отдели факты от моих прогнозов.
5. Сравни варианты условно.
6. Проведи pre-mortem для каждого.
7. Укажи информацию,
   которая сильнее всего может изменить выбор.

Не объявляй один вариант правильным,
если выбор зависит от моих ценностей.
```

## Обратимые и необратимые решения

```text
Раздели последствия решения на:

- обратимые;
- трудно обратимые;
- необратимые;
- неизвестные.

Какие маленькие действия позволят получить информацию,
не принимая сразу необратимое решение?
```

## Pre-mortem

```text
Представь, что через полгода выбранный вариант оказался ошибкой.

Назови пять разных правдоподобных причин.
Для каждой укажи:
- ранний сигнал;
- что можно проверить заранее;
- какое действие снизит риск.
```

ИИ структурирует решение. Ответственность и последствия остаются у человека.

# ИИ как учебный собеседник

Есть принципиальная разница между получением готового ответа и тренировкой собственного мышления.

## Режим замещения

```text
Реши задачу и дай готовый ответ.
```

Результат появился. Навык мог не появиться.

## Режим подсказок

```text
Не решай задачу за меня.

Сначала попроси предложить первый шаг.
Если я ошибусь,
укажи только первую существенную ошибку
и дай минимальную подсказку.

Полное решение покажи только после моей второй попытки.

Задача:
[вставить задачу]
```

## Проверка собственного решения

```text
Ниже условие и моё решение.

Не давай альтернативное решение сразу.

Проверь:
- соответствует ли ответ условию;
- где сделан первый неверный переход;
- какое правило нарушено;
- ошибка вычислительная или концептуальная.

После разбора дай похожую короткую задачу
именно на мою ошибку.
```

## Устный экзамен в чате

```text
Проверь, понимаю ли я тему
«корреляция и причинность».

Задавай по одному вопросу.
Не принимай правильный термин без объяснения
как доказательство понимания.

Используй новые примеры,
которых не было в нашем разговоре.

После пяти вопросов опиши конкретные пробелы
без общей похвалы.
```

## Проверка иллюзии понимания

```text
Я прочитал объяснение,
и мне кажется, что всё понятно.

Не пересказывай его снова.

Дай:
- новый пример;
- контрпример;
- задачу на перенос знания;
- вопрос «почему»,
  на который нельзя ответить определением.
```

Хорошее учебное применение увеличивает количество собственных попыток, обратной связи и переноса знаний. Плохое — заменяет именно ту умственную операцию, которой человек должен научиться.

# Итеративная редактура без деградации текста

Многократное «улучши ещё» постепенно меняет смысл. Надёжнее разделить виды редактирования.

## 1. Структура

```text
Найди в черновике ключевые мысли.

Предложи порядок блоков,
но пока не переписывай предложения.

Для каждого блока укажи его функцию в тексте.
```

## 2. Сокращение

```text
Удали повторы и фрагменты,
которые не добавляют нового смысла.

Не добавляй примеры и аргументы.
Покажи, что было удалено существенно.
```

## 3. Проверка аргумента

```text
Для каждого абзаца укажи:
- главную мысль;
- её основание;
- связь со следующим абзацем;
- нужен ли источник;
- можно ли удалить абзац без потери аргумента.
```

## 4. Стиль

```text
Проверь тон текста.

Аудитория: [описать].
Цель: [описать].
Желаемый тон: уверенный, но не высокомерный;
живой, но не фамильярный.

Укажи места, где тон нарушается.
Не переписывай текст целиком.
```

## 5. Сравнение версий

```text
Сравни исходную и новую версии.

Покажи:
- что стало яснее;
- какие факты потерялись;
- где позиция усилилась без основания;
- где изменился смысл;
- стала ли версия действительно лучше для заданной аудитории.
```

Сохраняйте удачную версию вне последовательности правок. Следующая итерация не обязана быть лучше предыдущей.

# Аудит всего диалога

В длинном разговоре полезно проверять не только последний ответ, но и путь, по которому к нему пришли.

```text
Проведи аудит нашего разговора.

Найди:
- мои предположения,
  которые позже стали использоваться как факты;
- твои утверждения без основания;
- потерянные ограничения;
- вопросы, на которые мы не ответили;
- решения, противоречащие друг другу;
- случаи согласия со мной без независимой проверки;
- места, где формулировка стала увереннее,
  хотя новых данных не появилось.

Не продолжай решать задачу.
Сначала покажи проблемы диалога.
```

## Восстановить чистое состояние

```text
На основе аудита собери чистое состояние задачи.

Включи только:
- цель;
- проверенные факты;
- действующие решения;
- ограничения;
- открытые вопросы;
- следующий проверяемый шаг.

Предположения помести в отдельный раздел.
Не включай историю обсуждения и отменённые варианты.
```

# Конфиденциальность остаётся важной даже в текстовом чате

Не вставляйте в публичный чат:

- пароли и ключи доступа;
- персональные данные;
- закрытую переписку;
- внутренние документы и коммерческие сведения;
- медицинские и финансовые данные;
- информацию, которую нельзя раскрывать по правилам организации.

Условия хранения и использования данных зависят от сервиса, тарифа, настроек и корпоративного договора. Обычное чат-окно нельзя автоматически считать защищённым хранилищем.

Простая замена имени на «человек А» не всегда анонимизирует текст: человека могут определить по редкому сочетанию должности, города, возраста и события.

Для учебных примеров используйте вымышленные или существенно обезличенные данные.

# Главная мысль второй части

Новичок пытается получить идеальный ответ одним промптом.

Опытный пользователь управляет процессом:

```text
уточняет задачу
→ разделяет её на этапы
→ показывает примеры
→ фиксирует состояние
→ создаёт независимые ветки
→ сталкивает генерацию с критикой
→ отделяет факты от предположений
→ завершает проверяемым действием
```

Сила чатового режима не только в том, что модель умеет написать ответ. Она в том, что человек может остановить разговор, изменить направление, проверить предпосылку, запросить противоположную позицию и не позволить красивому тексту незаметно заменить реальность.
