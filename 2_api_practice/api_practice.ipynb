```python
import json
import requests
from getpass import getpass
import xml.etree.ElementTree as ET

# =============================================
# НАСТРОЙКИ API
# =============================================
API_URL = "http://127.0.0.1:8015/v1/chat/completions"
MODEL = "Qwen3.6-35B"
API_KEY = getpass("API-ключ (оставьте пустым для локального сервера): ")

HEADERS = {"Content-Type": "application/json"}
if API_KEY:
    HEADERS["Authorization"] = f"Bearer {API_KEY}"

# =============================================
# ОСНОВНАЯ ФУНКЦИЯ
# =============================================


def chat(messages, *args, **options):
    """Отправить запрос к Chat Completions API."""
    body = {"model": MODEL, "messages": messages, "stream": False}
    body.update(options)
    if body.get("stream"):
        raise ValueError("Эта функция рассчитана на stream=False")

    print(body)
    
    response = requests.post(
        API_URL,
        headers=HEADERS,
        json=body,
        timeout=60,
    )
    response.raise_for_status()
    return response.json()

# =============================================
# ФУНКЦИЯ ПАРСИНГА XML
# =============================================
def load_cases(xml_path):
    """Загрузить кейсы из XML-файла."""
    tree = ET.parse(xml_path)
    root = tree.getroot()
    
    cases = []
    for case in root.findall('case'):
        case_data = {
            'id': int(case.find('id').text),
            'input': case.find('input').text,
        }
        expected_el = case.find('expected')
        if expected_el is not None:
            case_data['expected'] = {child.tag: child.text for child in expected_el}
        cases.append(case_data)
    
    return cases

print("✅ Все функции готовы к работе!")
print(f"   Модель: {MODEL}")
print(f"   API URL: {API_URL}")
```
