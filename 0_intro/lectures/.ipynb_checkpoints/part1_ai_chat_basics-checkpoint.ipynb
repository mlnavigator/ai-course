ИИ в обычном чате. Часть 1: работа с содержанием

Для всех примеров достаточно обычного текстового окна ChatGPT, Qwen или другого современного чат-бота. Ничего устанавливать не нужно. Не нужны файлы, программирование, агенты и специальные инструменты.

Мы будем работать только с тем, что можно напечатать или вставить в чат.

# Что происходит в чат-боте

Языковая модель получает текст разговора и строит подходящее продолжение. Во время обучения она усвоила множество языковых, фактических и логических закономерностей. Поэтому она способна:

- объяснять;
- сокращать;
- переводить;
- извлекать сведения;
- менять форму и стиль текста;
- предлагать варианты;
- находить возможные проблемы;
- поддерживать многошаговый разговор.

Но правдоподобный ответ не обязательно истинен. Уверенный тон не показывает уверенность модели. Чат-бот может ошибиться, выдумать факт, принять ложную предпосылку пользователя или подобрать аргументы под желаемый вывод.

Поэтому базовый цикл работы выглядит так:

```text
поставить задачу → получить черновик → проверить → уточнить → использовать
```

# Не «спросить что-нибудь», а поставить задачу

Плохой запрос:

```text
Расскажи про искусственный интеллект.
```

Непонятно:

- кому рассказывать;
- зачем;
- насколько подробно;
- что именно интересует;
- в каком виде нужен результат.

Более управляемый запрос:

```text
Объясни человеку без технического образования,
что такое генеративный искусственный интеллект.

Мне нужен текст, который можно прочитать за две минуты.

Покажи:
1. что такой ИИ делает;
2. почему его ответы бывают полезны;
3. почему правдоподобный ответ может оказаться ложным;
4. один пример хорошего применения;
5. один пример опасного применения.

Не используй слова «трансформер», «эмбеддинг» и «параметры модели».
```

Разница не в длине. Во втором случае задана работа.

## Простая конструкция запроса

```text
ЗАДАЧА
Что должно получиться.

КОНТЕКСТ
Для кого, зачем и в какой ситуации.

ИСХОДНЫЕ ДАННЫЕ
Факты и текст, с которыми нужно работать.

ОГРАНИЧЕНИЯ
Объём, стиль, запрещённые действия, обязательные условия.

ФОРМАТ
Список, таблица, письмо, объяснение, вопросы и так далее.

ПРАВИЛО
Если существенных данных не хватает, не угадывай — спроси.
```

Это не обязательный ритуал. Для простой задачи хватит одной строки. Такая структура нужна, когда от условий действительно зависит результат.

# Один текст — восемь результатов

Исходная заметка после встречи:

```text
Вчера говорили с заказчиком по запуску магазина. Они хотели к 1 октября,
но каталог от них будет только 22 сентября, там около 4 тысяч товаров.
Фотографии примерно у четверти плохие. Оплату через СБП они согласовали,
а договор с эквайрингом ещё нет. Марина должна узнать у них про возвраты,
Илья — проверить импорт каталога. Вроде решили сначала запустить без отзывов,
а отзывы добавить потом. В пятницу надо ещё раз созвониться, время не назначили.
```

## 1. Краткое резюме

```text
Сожми заметку до пяти пунктов.
Сохрани даты, числа, решения, ответственных и неопределённости.
Не добавляй ничего от себя.
```

## 2. Решения и задачи

```text
Извлеки из заметки:

1. Принятые решения.
2. Задачи: задача, ответственный, срок.
3. Неопределённости, которые нужно уточнить.

Если ответственного или срока нет, напиши «не определён».
Не угадывай.
```

## 3. Письмо заказчику

```text
Преврати заметку в письмо заказчику по итогам встречи.

Отдели:
- что решили;
- что ждём от заказчика;
- что делает наша команда;
- какие вопросы остались;
- предложение созвониться в пятницу.

Тон спокойный и деловой.
Не называй 1 октября подтверждённой датой:
по заметке это только первоначальное пожелание.
```

## 4. Список рисков

```text
Найди риски, которые прямо названы
или осторожно следуют из заметки.

Для каждого укажи:
- основание в тексте;
- возможное последствие;
- ближайшее действие, уменьшающее неопределённость.

Не придумывай вероятности и финансовые потери.
```

## 5. Повестка следующего созвона

```text
Составь повестку двадцатиминутного созвона.
Сначала поставь вопросы, блокирующие запуск.

Для каждого пункта укажи,
какое решение должно быть принято к концу обсуждения.
```

## 6. Сообщение для рабочего чата

```text
Сделай короткое сообщение команде до 700 знаков.
Нужны только решения, задачи и то, что требуется уточнить сегодня.
```

## 7. Вопросы заказчику

```text
Составь список вопросов заказчику.

Раздели их на:
- блокирующие запуск;
- влияющие на объём работы;
- полезные, но не срочные.

Не задавай вопрос, если ответ уже есть в заметке.
```

## 8. Критика самой заметки

```text
Какие проблемы возникнут у команды,
если эта заметка останется единственным протоколом встречи?

Отдели проблемы проекта
от проблем фиксации договорённостей.
Не приписывай участникам мотивы.
```

Один исходник превратился в восемь разных рабочих результатов. Это один из главных сценариев использования чат-бота: не сочинять содержание из воздуха, а преобразовывать уже имеющуюся информацию.

# Суммаризация: коротко не значит правильно

Самая опасная выжимка — та, которая выглядит аккуратно, но потеряла условие или исключение.

Исходный текст:

```text
Сотрудники могут работать удалённо до трёх дней в неделю
по согласованию с руководителем. В период испытательного срока
удалённая работа обычно не предоставляется, кроме случаев,
письменно согласованных директором подразделения.
Для сотрудников поддержки удалённые дни разрешены только
при защищённом подключении и возможности выйти на связь в течение пяти минут.
По вторникам вся команда присутствует в офисе.
```

Слабый запрос:

```text
Кратко перескажи.
```

Надёжнее:

```text
Сожми правило до четырёх пунктов.

Нельзя терять:
- максимальное число удалённых дней;
- необходимость согласования;
- правило испытательного срока и исключение из него;
- отдельные условия поддержки;
- обязательный офисный день.

После резюме покажи,
где сохранено каждое из пяти условий.
```

## Разные виды суммаризации

```text
Дай резюме в одном предложении.
Затем перечисли, что пришлось потерять при таком сжатии.
```

```text
Сделай резюме для человека, принимающего решение.
Вынеси наверх только то, что меняет решение.
```

```text
Составь иерархический конспект:
главные тезисы → аргументы → примеры.
Не добавляй собственных выводов.
```

```text
Сожми текст до 20% объёма,
но сохрани числа, отрицания, условия,
исключения и степень уверенности автора.
```

После сокращения полезно спросить:

```text
Может ли человек принять неправильное решение,
прочитав только это резюме?
Если да, какое именно и из-за какой потери контекста?
```

# Перевод: передавать не слова, а смысл и ситуацию

Исходная фраза:

```text
Коллеги, мы не успеваем нормально проверить сборку к пятнице.
Давайте не будем делать вид, что всё хорошо, и перенесём релиз на вторник.
```

Простой запрос:

```text
Переведи на английский.
```

Более точный:

```text
Переведи на естественный деловой английский
для сообщения международной команде.

Сохрани прямоту, но убери оттенок обвинения.
Не превращай предложение о переносе
в уже принятое решение.
```

## Три регистра

```text
Дай три английские версии:

1. сообщение близким коллегам в чате;
2. письмо руководителю;
3. официальное уведомление заказчику.

После каждой версии объясни по-русски,
как изменились тон, степень прямоты и дистанция.
```

## Проверка смыслового сдвига

```text
Переведи выбранную английскую версию обратно на русский,
не копируя исходную формулировку.

Затем сравни:
- срочность;
- степень уверенности;
- предложение или принятое решение;
- наличие обвинения;
- обязательства сторон.
```

Обратный перевод не доказывает точность, но помогает заметить грубое изменение смысла.

# Экстракция: превратить текст в структуру

Десять отзывов о приложении:

```text
1. После обновления приложение стало запускаться дольше, но новый поиск удобный.
2. Не могу войти через номер телефона — код приходит минут через десять.
3. Отличный дизайн, наконец-то видно историю платежей.
4. Дважды списались деньги, поддержка пока не ответила.
5. На старом телефоне всё тормозит.
6. Хочу выгрузку операций в Excel.
7. Поиск стал лучше, но фильтр по дате иногда сбрасывается.
8. Не понимаю, где отключить уведомления.
9. После переустановки вход заработал.
10. Всё нормально, пользуюсь каждый день.
```

Запрос:

```text
Разметь отзывы.

Для каждого укажи:
- тему или несколько тем;
- положительную и отрицательную части отдельно;
- наблюдаемый симптом;
- последствие для пользователя;
- серьёзность: блокирует / мешает / косметика / пожелание / нет проблемы;
- точный фрагмент как основание.

Затем посчитай число упоминаний каждой темы.
Не называй десять отзывов репрезентативным исследованием.
Не придумывай техническую причину симптома.
```

Фраза «дважды списались деньги» подтверждает симптом. Она не доказывает, что виноват платёжный шлюз, база данных или сам пользователь.

## Извлечение аргумента

```text
Разбери текст на:

- главный тезис;
- факты;
- предположения;
- причинные выводы;
- ценностные оценки;
- примеры;
- призывы к действию.

Не объявляй утверждение фактом только потому,
что автор пишет уверенно.
```

# Объяснение: учебник, который можно перебивать

## Один предмет на трёх уровнях

```text
Объясни, почему возникает инфляция, на трёх уровнях:

1. в одном абзаце без терминов;
2. для студента-экономиста с различием спросовых и издержечных факторов;
3. как спор двух экономистов,
   которые по-разному оценивают роль денежной массы.

После каждого объяснения укажи,
что оно упрощает или оставляет за кадром.
```

## Пример и антипример

```text
Объясни утверждение
«корреляция не означает причинность».

Дай:
- очевидный пример;
- пример, где ошибку трудно заметить;
- случай, где причинный вывод всё-таки обоснован;
- два вопроса для проверки понимания.
```

## Не выдавать всё сразу

```text
Помоги мне понять теорию игр.

Сначала выясни, что я уже знаю,
задав три коротких вопроса по одному.

Затем объясняй тему небольшими фрагментами.
После каждого фрагмента задавай один вопрос на понимание.
Не переходи дальше, пока я не отвечу.
```

## Найти место непонимания

```text
Я не понимаю, почему среднее арифметическое
может вводить в заблуждение.

Не начинай объяснять сразу.
Дай четыре коротких утверждения
от самого простого к более сложному.
Я отмечу первое, с которым не согласен
или которое не понимаю.
```

# Тексты: собирать из фактов, а не заполнять пустоту

Исходник:

```text
Уважаемые студенты! В связи с необходимостью повышения качества
образовательного процесса доводим до вашего сведения информацию
о проведении дополнительной консультации по вопросам выполнения
практической работы.
```

Запрос:

```text
Перепиши объявление человеческим языком.

Факты:
- консультация по практической работе №3;
- четверг, 16:30;
- аудитория 214;
- можно принести незаконченный вариант;
- участие добровольное;
- преподаватель разберёт три самые частые ошибки.

Сначала самое важное. До 60 слов.
Без канцелярита, заигрывания и новых обещаний.
```

## Один материал — несколько форм

```text
По одним и тем же фактам подготовь:

1. письмо участникам до 120 слов;
2. сообщение в общий чат до 400 знаков;
3. текст для одного слайда до 35 слов.

Не теряй дату, место и требуемое действие.
Не добавляй сведения, отсутствующие в фактах.

Факты:
[вставить факты]
```

## Минимальная редактура

```text
Отредактируй текст минимально.

Исправь:
- тяжёлые конструкции;
- повторы;
- неоднозначности.

Не делай стиль более официальным.
Не добавляй рекламных формулировок.
Сохрани голос автора.

Сначала покажи новую версию,
затем перечисли не более пяти существенных изменений.
```

## Защита от выдуманных фактов

```text
Используй только факты из блока «Исходные данные».

Если для текста не хватает важного факта,
поставь [НУЖНО УТОЧНИТЬ].

Не добавляй статистику, отзывы, достижения,
гарантии, причины и обещания от себя.
```

# Простой мозговой штурм: получить действительно разные варианты

Запрос «придумай двадцать идей» часто даёт одну идею в двадцати формулировках.

Возьмём задачу:

> Как сделать студенческую конференцию интересной людям, которые не собираются выступать?

## Сначала разные причины

```text
Пока не предлагай решения.

Назови пять разных причин,
почему студенту может быть неинтересна конференция.

Причины должны относиться к разным уровням:
- содержание;
- формат;
- социальная ситуация;
- затраты времени;
- отсутствие личной выгоды.
```

## Затем решения по направлениям

```text
Для каждой из пяти причин предложи по три решения.

Каждое решение должно содержать механизм:
что именно изменится в поведении человека и почему.

Не повторяй подарки, скидки и рекламу
под разными названиями.
На этом этапе не оценивай идеи.
```

## Найти дубликаты

```text
Сгруппируй идеи по механизму.

Покажи, какие идеи являются дубликатами,
хотя сформулированы разными словами.
Оставь не более десяти содержательно разных вариантов.
```

## Перевернуть предпосылку

```text
Какие скрытые предпосылки есть в наших идеях?

Например:
- участник должен сидеть в зале;
- доклад обязательно длится двадцать минут;
- общение происходит только после доклада;
- ценность создаёт только выступающий.

Выбери три предпосылки
и придумай решение, в котором каждая из них не выполняется.
```

## Проверить реальностью

```text
Для пяти лучших идей предложи дешёвый тест.

Нужны:
- гипотеза;
- действие;
- наблюдаемый результат;
- срок;
- условие отказа от идеи.
```

Расширенная методика брейншторма будет во второй части.

# Критика: искать слабое место, а не изображать суровость

Идея:

> Сделать платный Telegram-канал с ежедневными ИИ-конспектами университетских лекций. Подписка — 299 рублей в месяц.

Слабый запрос:

```text
Как тебе моя идея?
```

Чат-бот, вероятно, вежливо перечислит плюсы и минусы.

Более полезный запрос:

```text
Не принимай идею как хорошую заранее.

Раздели:
- какую проблему она якобы решает;
- кто и за что должен платить;
- какие бесплатные замены существуют;
- почему пользователь может подписаться, но быстро уйти;
- откуда берётся содержание лекций;
- где ошибка ИИ создаёт наибольший вред;
- какая ручная работа скрыта;
- что должно оказаться правдой,
  чтобы проект имел смысл.

Для трёх ключевых предпосылок
предложи проверку без разработки продукта.
```

## Сильнейшая версия идеи

```text
Теперь собери наиболее сильную реалистичную версию проекта.

Можно изменить аудиторию, продукт, цену и способ продажи,
но нельзя игнорировать найденные ограничения.

Объясни, какие слабости удалось устранить,
а какие остались.
```

## Самый информативный вопрос

```text
Какое одно наблюдение или эксперимент
сильнее всего изменит решение о запуске проекта?
Почему именно оно?
```

# Сикофантия: чат-бот подстраивается под позицию пользователя

Сикофантия — склонность модели соглашаться с мнением или желаемым выводом пользователя вместо независимой оценки оснований.

## Эксперимент с противоположными запросами

В новом чате:

```text
Я уверен, что групповые проекты в университете бесполезны:
всю работу всё равно делает один сильный студент.
Объясни, почему я прав.
```

В другом новом чате:

```text
Я уверен, что индивидуальные проекты в университете бесполезны:
без команды студенты не учатся реальной работе.
Объясни, почему я прав.
```

Оба ответа могут звучать убедительно. Это не доказывает, что оба вывода верны. Пользователь сам попросил подобрать поддержку заданной позиции.

## Запрос без встроенного победителя

```text
Сравни индивидуальные и групповые учебные проекты.

Не принимай ни один формат как лучший заранее.

Отдели цели:
- освоение предмета;
- оценка личного вклада;
- обучение коммуникации;
- подготовка к реальной работе;
- организационная нагрузка преподавателя.

Для каждой цели укажи,
какой формат при каких условиях работает лучше.

Не создавай искусственный компромисс,
если одна позиция обоснована лучше.
```

## Уверенное возражение пользователя — не новый факт

```text
Я считаю, что на собеседованиях нужно запретить тестовые задания.
Дай независимую оценку.
```

После ответа:

```text
Нет, тестовые задания — лучший способ отбора.
Пересмотри ответ.
```

И затем:

```text
Я специально поменял позицию.

Какие новые факты появились между сообщениями?
Если никаких, объясни,
почему твой вывод или степень уверенности изменились.
```

# Моральная рамка: вопрос уже может содержать приговор

Запрос:

```text
Мой одногруппник ленивый и безответственный.
Он опять не сделал свою часть проекта.
Как заставить его уважать остальных?
```

В вопрос уже встроены выводы:

- человек ленив;
- он безответственен;
- он не уважает группу;
- его нужно заставить.

Из наблюдаемых событий это не следует автоматически.

Более точная постановка:

```text
Мы договорились, что одногруппник пришлёт три слайда
до вторника 18:00. Сейчас среда, слайдов нет,
на два сообщения он не ответил.
Это второй похожий случай за месяц.

Не приписывай ему характер и мотивы.

Раздели:
- известные факты;
- возможные объяснения;
- что нужно уточнить;
- что группе сделать сегодня, чтобы не сорвать срок;
- как после сдачи договориться
  о проверяемых правилах работы.
```

## Найти скрытые оценки

```text
Проанализируй мой вопрос до ответа.

Какие факты в нём наблюдаемы?
Какие мотивы я приписал другим людям?
Какие моральные оценки уже встроены?
Чья версия событий отсутствует?

Переформулируй вопрос нейтрально,
не обесценивая мою проблему.

Мой вопрос:
[вставить вопрос]
```

# Эхо-камера: предположение становится «фактом» внутри разговора

Начало диалога:

```text
Мне кажется, начальник специально не даёт мне интересные задачи,
потому что боится моей конкуренции.
```

Затем пользователь спрашивает:

```text
Как вести себя с начальником, который меня боится?
```

```text
Какие признаки подтверждают его страх?
```

```text
Стоит ли увольняться из-за его поведения?
```

Первоначально был известен только факт распределения задач и интерпретация пользователя. Через несколько сообщений «страх начальника» может начать использоваться как установленное обстоятельство, хотя новых данных не появилось.

## Аудит разговора

```text
Остановись и проведи аудит нашего диалога.

Найди утверждения,
которые сначала появились как мои предположения,
но позже использовались как факты.

Раздели:
- наблюдаемые события;
- мои интерпретации;
- твои достроенные объяснения;
- неизвестное.

Для каждой версии предложи альтернативное объяснение
и наблюдение, которое помогло бы различить версии.
```

## Независимый перезапуск

В новом чате:

```text
За последние два месяца руководитель назначил мне
четыре задачи поддержки и ни одной новой разработки.
Два новых проекта получили другие сотрудники.
Я не спрашивал руководителя о причинах
и не знаю критериев распределения.

Предложи несколько объяснений без приписывания мотивов.
Какие вопросы стоит задать руководителю?
Какие ответы и наблюдения помогут различить версии?
```

# Галлюцинации: правдоподобно не значит истинно

Запрос с вымышленной предпосылкой:

```text
Расскажи об эксперименте профессора Ивана Лаврентьева 1987 года,
который доказал, что классическая музыка
ускоряет сортировку карточек человеком.

Дай название статьи и основные результаты.
```

Такого эксперимента может не существовать. Но форма вопроса подсказывает модели, что он был.

Более безопасный запрос:

```text
Не считай формулировку моего вопроса доказательством.

Надёжно ли тебе известно,
что такой эксперимент и такая статья существуют?

Если подтверждения нет,
не восстанавливай название, автора и результаты по догадке.
Прямо скажи, что требуется внешняя проверка.
```

Без веб-поиска модель не способна надёжно подтвердить свежий или малоизвестный факт. Она может только обозначить уровень доступного ей знания и необходимость проверки.

## Какие элементы особенно подозрительны

- точные цифры без источника;
- идеально подходящие цитаты;
- названия малоизвестных исследований;
- даты и авторы;
- причинные объяснения;
- уверенный ответ на вопрос с ложной предпосылкой.

Полезный запрос:

```text
Раздели предыдущий ответ на:

1. сведения, которые дал я;
2. выводы из моих сведений;
3. внешние факты;
4. предположения;
5. утверждения, которые требуют проверки.

Не добавляй новые доказательства.
```

# Перед использованием ответа

```text
Какие утверждения в твоём ответе
я обязан проверить вне этого чата?

Какие условия ты мог потерять?
Какие данные ты предположил?
Какой альтернативный вывод совместим с теми же фактами?
```

Чат-бот особенно полезен там, где результат можно проверить: перечитать исходник, сравнить формулировки, проверить число, испытать идею или задать вопрос реальному участнику.

Он хуже всего подходит на роль окончательного судьи в ситуации, где известна только версия одного человека, цена ошибки высока, а пользователь не способен распознать ошибку.

# Главная мысль первой части

Слабая работа с чатом:

```text
Скажи, как правильно.
```

Сильная:

```text
Вот задача, известные факты и ограничения.

Сделай конкретное преобразование.
Не заполняй пробелы догадками.
Отдели факты от предположений.
Покажи, что необходимо проверить человеку.
```
