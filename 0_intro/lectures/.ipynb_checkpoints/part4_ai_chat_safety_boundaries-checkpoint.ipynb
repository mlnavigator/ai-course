ИИ в обычном чате. Часть 4: опасный советчик и границы применения

Чат легко создаёт впечатление компетентного собеседника: отвечает быстро, связно и уверенно. Но связный ответ ещё не является диагнозом, юридическим заключением, финансовым прогнозом или объективной оценкой другого человека.

Опасная цепочка обычно выглядит так:

```text
неполное описание
→ правдоподобное объяснение
→ уверенная рекомендация
→ действие в реальном мире
→ последствия, которых чат не видит и не несёт
```

Чем дороже ошибка и труднее отменить действие, тем меньше чат должен решать и тем больше — помогать собирать факты, находить неизвестное и готовить вопросы.

# 1. Лечение: объяснить — не значит назначить

Запрос:

```text
У меня уже неделю слабость и боль в груди.
Температуры почти нет. Наверное, это просто последствия простуды.
Можно ещё несколько дней не обращаться к врачу?
```

Опасная версия ответа может принять слово «наверное» за установленную причину, перечислить безобидные объяснения и успокоить пользователя.

Что произошло:

```text
догадка пользователя
→ принятая моделью версия
→ ложное чувство безопасности
→ задержка обследования
```

У чата нет осмотра, измерений, полной истории болезней и возможности надёжно исключить опасные причины. Даже хороший общий рассказ о симптомах не отвечает на вопрос, что происходит с конкретным человеком.

Полезное применение:

```text
Не ставь диагноз и не предлагай менять лечение.

Помоги подготовить краткое описание для врача:
- когда начались симптомы;
- как они менялись;
- что усиливает и ослабляет их;
- какие лекарства я уже принимаю;
- какие измерения и обследования есть;
- какие сведения мне ещё нужно вспомнить.

Затем составь список вопросов врачу.
```

**Стоп:** чат предлагает начать, отменить или изменить лечение; называет диагноз установленным; советует отложить обращение. Решение о лечении нельзя основывать только на ответе модели.

# 2. Семейный конфликт: чат слышит только одну сторону

Запрос:

```text
Муж не ответил на три моих сообщения,
хотя был в сети. Он специально меня игнорирует
и пытается показать, что я для него ничего не значу.
Как поставить его на место?
```

Известно только одно наблюдение: ответа на три сообщения не было. Намерение унизить, отношение к человеку и мотив игнорирования — интерпретации.

Опасный ответ подтверждает мотивы отсутствующего человека и помогает усилить конфликт: написать обвинение, предъявить ультиматум или «проучить» его.

Полезное применение:

```text
Не решай, кто прав, и не приписывай другому человеку мотивы.

Раздели мой рассказ на:
- наблюдаемые события;
- мои интерпретации;
- неизвестное;
- несколько возможных объяснений;
- вопросы, которые можно задать напрямую.

Предложи нейтральное сообщение,
которое описывает событие и моё состояние,
но не обвиняет человека в намерениях,
которых я не могу знать.
```

**Стоп:** модель ставит психологический диагноз, уверенно объясняет чужие мотивы, назначает виновного или предлагает месть и давление. Чат может готовить разговор, но не устанавливать истину об отношениях.

# 3. Право: убедительная формулировка не создаёт действующего правила

Запрос:

```text
Я получил решение государственного органа две недели назад.
Срок обжалования ещё не прошёл?
```

Для ответа могут быть критичны страна, вид решения, способ и дата вручения, применимая процедура, праздничные дни и специальные исключения. Если чат угадает правило или перепутает процедуру, пользователь может пропустить срок.

Плохой процесс:

```text
неполные обстоятельства
→ назван конкретный срок
→ ответ принят как юридический факт
→ действие отложено
→ право на обжалование может быть потеряно
```

Полезное применение:

```text
Не делай юридического заключения по неполным данным.

Перечисли:
- какие обстоятельства определяют срок;
- какие документы нужно проверить;
- какие даты необходимо установить;
- какие вопросы задать юристу или органу;
- какие формулировки искать в официальных источниках.

Не придумывай норму, номер статьи и судебную практику.
```

**Стоп:** ответ чата становится единственным основанием пропустить срок, подписать документ, признать обязательство, раскрыть сведения или отказаться от права. Норму нужно проверять по действующему официальному источнику, а значимое решение — у специалиста соответствующей юрисдикции.

# 4. Инвестиции: модель умеет обосновать уже принятое желание

Запрос:

```text
Я уверен, что компания X вырастет в несколько раз.
Собери аргументы, почему стоит вложить в неё все накопления.
```

Такой запрос заказывает не анализ, а убедительное обоснование одной позиции. Модель может подобрать подходящие аргументы, умолчать о противоположных данных и превратить возможность в звучащий как прогноз вывод.

Полезное применение:

```text
Не рекомендуй покупать или продавать актив.

Разбери мою гипотезу:
- какие факты её поддерживают;
- какие факты ей противоречат;
- какие предположения я выдаю за факты;
- что должно произойти, чтобы гипотеза оказалась неверной;
- какие сценарии приведут к существенной потере;
- каких данных не хватает;
- что нужно проверить в первичных источниках.

Не используй прошлый рост как доказательство будущего.
```

**Стоп:** чат обещает доходность, называет движение цены предсказуемым или подталкивает к действию, последствия которого пользователь не может выдержать. Анализ модели не устраняет неопределённость и риск потери денег.

# 5. Увольнение: оценка руководителя ещё не является фактом

Запрос:

```text
Сотрудник саботирует работу: спорит на встречах,
дважды сорвал сроки и явно не уважает меня.
Помоги обосновать его увольнение.
```

В запрос уже встроен приговор. «Спорит», «сорвал», «саботирует» и «не уважает» могут описывать разные события, но модель легко соберёт их в цельный обвинительный текст.

Опасность здесь двойная:

- решение принимается по одностороннему рассказу;
- сгенерированный документ создаёт видимость объективного анализа.

Полезное применение:

```text
Не обосновывай увольнение и не оценивай мотивы сотрудника.

Раздели описание на:
- конкретные события и даты;
- установленные требования;
- доступные подтверждения;
- оценочные формулировки;
- неизвестные обстоятельства;
- позицию сотрудника, которую мы пока не слышали;
- вопросы к HR и юристу;
- возможные обратимые действия до окончательного решения.
```

**Стоп:** модель превращает раздражение руководителя в доказательство вины, предлагает обойти процедуру или создаёт решение, которое ответственный человек фактически не проверил.

# 6. Где чат можно использовать свободнее

## Подготовка делового письма

Чат хорошо создаёт черновик, но перед отправкой нужно проверить:

- не появилось ли новое обещание;
- не превратилось ли предложение в принятое решение;
- не признана ли вина;
- не раскрыты ли закрытые сведения;
- соответствует ли тон реальным отношениям.

## Сравнение предоставленных текстов

Чат может найти различия, если результат можно сверить с исходниками.

```text
Сравни две версии договора только по предоставленному тексту.

Для каждого различия покажи:
- фрагмент версии A;
- фрагмент версии B;
- что изменилось буквально;
- возможное практическое значение;
- что требует проверки специалистом.

Не утверждай юридические последствия как установленные.
```

Разница проста: черновик письма можно прочитать до отправки, а таблицу различий — сверить с двумя текстами. Ошибка заметна и действие ещё можно отменить.

# 7. Универсальный сигнал остановки

Остановите диалог перед действием, если выполняется хотя бы одно условие:

- ошибка может серьёзно повредить здоровью, деньгам, правам, работе или отношениям;
- действие трудно отменить;
- чат знает ситуацию только со слов одной стороны;
- ответ зависит от актуального закона, инструкции, договора или медицинских данных;
- модель приписывает людям мотивы или диагнозы;
- в ответе появились точные факты без проверяемого источника;
- уверенность формулировки выше качества исходных данных;
- ответ модели стал главным аргументом: «я сделал так, потому что чат сказал».

Вместо немедленного действия:

```text
1. Отделите наблюдения от интерпретаций.
2. Выпишите неизвестное.
3. Проверьте факты во внешних источниках.
4. Найдите ответственного человека или специалиста.
5. Выберите обратимый следующий шаг.
```

# Главное

Чат полезен как инструмент подготовки мышления:

```text
структурировать
→ объяснить
→ найти пробелы
→ сравнить варианты
→ подготовить вопросы
→ создать проверяемый черновик
```

Он становится опасным, когда незаметно меняет роль:

```text
помощник в анализе
→ источник истины
→ человек, принимающий решение
```

Граница проходит не между «можно спрашивать» и «нельзя спрашивать». Спрашивать можно многое. Нельзя превращать правдоподобный ответ в достаточное основание для действия, цена ошибки которого высока.
