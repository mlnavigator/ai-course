ИИ в работе с текстом. Часть 3: исследование, синтез и редактура

Хороший текст редко появляется из запроса «напиши статью». Обычно работа выглядит так:

```text
вопрос
→ источники
→ анализ
→ сравнение
→ синтез
→ структура
→ черновик
→ редактура
→ адаптация
→ проверка
```

ИИ может ускорить каждый этап. Но он не превращает непроверенные сведения в факты и не снимает с автора ответственность за итоговый текст.

# 1. Сначала вопрос, потом текст

Плохое начало:

```text
Напиши статью о влиянии удалённой работы на карьеру.
```

Здесь не определены группа, период, вид удалённой работы и само понятие карьеры. Модель заполнит пробелы наиболее привычными рассуждениями.

Лучше начать так:

```text
Помоги превратить тему
«влияние удалённой работы на карьеру»
в исследовательский вопрос.

Уточни:
- какую группу работников рассматриваем;
- что понимаем под карьерным ростом;
- с чем сравниваем удалённую работу;
- какой период рассматриваем;
- какие результаты можно наблюдать;
- какие факторы способны изменить эффект.

Пока не пиши статью.
```

Допустимый результат:

> Как полностью удалённая работа влияет на скорость повышения, рост дохода и доступ к наставничеству у специалистов в первые пять лет карьеры по сравнению с гибридной работой?

Узкий вопрос не гарантирует правильного ответа, но позволяет понять, какие данные вообще нужны.

# 2. Как получать исходный материал

Есть три режима.

## Обычный чат

Подходит для:

- уточнения вопроса;
- построения гипотез;
- объяснения терминов;
- разработки структуры;
- обработки уже вставленных текстов.

Он не подтверждает актуальные или малоизвестные факты.

## Веб-поиск

Подходит, когда нужно быстро найти конкретный документ, правило, цифру или официальный источник.

## Deep Research

Подходит для вопроса, где нужно сделать много поисковых шагов, сопоставить разные источники и собрать отчёт со ссылками.

Пример задания:

```text
Исследуй влияние удалённой работы
на карьерное развитие молодых специалистов.

Рассматривай отдельно:
- повышение в должности;
- рост дохода;
- доступ к наставничеству;
- получение сложных задач;
- профессиональные связи;
- полностью удалённую и гибридную работу.

Не объединяй в один вывод разные профессии,
страны, периоды и определения удалённой работы.

Приоритет источников:
1. систематические обзоры;
2. рецензируемые эмпирические исследования;
3. официальная статистика;
4. исследования с раскрытой методологией;
5. экспертные мнения — только как интерпретация.

Для каждого существенного вывода дай источник.
Покажи подтверждающие и противоречащие данные.

Отдельно укажи:
- что установлено достаточно надёжно;
- что зависит от условий;
- что пока неизвестно;
- где есть только корреляция.
```

Если поиск недоступен, можно вставить в чат несколько текстов или выдержек. Все следующие этапы остаются теми же.

# 3. Разобрать каждый источник отдельно

До объединения источников полезно сделать их карточки.

```text
Для каждого источника составь карточку:

- название или обозначение;
- дата;
- тип материала;
- исследовательский вопрос;
- изучаемая группа;
- использованные данные;
- метод;
- главный результат;
- ограничения;
- что источник способен подтвердить;
- чего он подтвердить не может.

Не оценивай источник только по уверенности языка.
```

Так сложнее случайно перенести результат об опытных разработчиках на всех молодых специалистов или выдать опрос мнений за измерение реального поведения.

# 4. Сравнить несколько источников

```text
Сравни источники по таблице:

источник |
вопрос |
группа и период |
определение удалённой работы |
показатель карьерного результата |
метод |
основной результат |
ограничения.

После таблицы покажи:
- в чём источники согласны;
- где результаты зависят от условий;
- где есть настоящее противоречие;
- какие вопросы остаются без ответа.
```

## Мнимое противоречие

Два утверждения могут выглядеть несовместимыми:

- «Удалённая работа повышает производительность».
- «Удалённая работа снижает производительность».

Но первое исследование может изучать индивидуальную аналитическую работу опытных сотрудников, а второе — обучение новичков в команде.

Проверка:

```text
Для каждой пары противоречащих утверждений проверь:

- одинаковый ли вопрос рассматривается;
- одинаково ли определены термины;
- одинаковы ли группы и периоды;
- одинаковы ли измеряемые результаты;
- различаются ли условия.

Верни:
утверждение A | утверждение B |
реальное противоречие или разный контекст |
объяснение | что нужно проверить.
```

# 5. Связать утверждения с доказательствами

Реальная ссылка рядом с абзацем ещё не означает, что она подтверждает написанное.

```text
Построй таблицу:

утверждение |
источник |
подтверждающий фрагмент |
прямое утверждение или наш вывод |
ограничение |
насколько сильную формулировку можно использовать.

Если источник относится только к общей теме,
но не подтверждает конкретное утверждение,
пометь «не подтверждает».
```

Особенно проверяются:

- числа;
- цитаты;
- причинные связи;
- обобщения на другие группы;
- слова «доказано», «всегда», «приводит», «обеспечивает».

# 6. Синтезировать, а не склеивать пересказы

Слабый обзор устроен так:

```text
Автор A сказал...
Автор B сказал...
Автор C сказал...
```

Синтез организует материал по вопросам и результатам:

```text
На основе карточек источников построй синтез.

Не пересказывай источники по очереди.

Организуй результат по вопросам:
1. В чём источники согласны?
2. Какие результаты зависят от условий?
3. Где есть реальные противоречия?
4. Какие механизмы предлагаются?
5. Какие данные наиболее применимы к нашему вопросу?
6. Какие выводы делать нельзя?
7. Что остаётся неизвестным?

Каждое эмпирическое утверждение свяжи с источником.
Не усредняй несовместимые результаты.
```

# 7. Создать опорный материал

До написания статьи полезно зафиксировать проверенное содержание.

```text
Собери опорный материал для будущего текста.

Разделы:
- исследовательский вопрос;
- рабочий тезис;
- подтверждённые факты;
- условные выводы;
- противоречащие данные;
- возможные объяснения;
- сильнейшее возражение;
- ограничения;
- неизвестное;
- допустимые источники;
- утверждения, которые нельзя использовать.

Не пиши статью.
```

Этот документ становится источником истины для черновика и дальнейших форматов.

# 8. Спроектировать структуру

```text
Предложи три структуры текста:

1. проблема → причины → последствия → решения;
2. тезис → доказательства → возражение → уточнённый вывод;
3. конкурирующие объяснения → проверка → синтез.

Для каждой укажи:
- какую читательскую задачу она решает;
- где находится главный тезис;
- где разместить противоречащие данные;
- какое место будет самым слабым;
- для какой аудитории структура подходит.

Не пиши сам текст.
```

После выбора:

```text
Напиши черновик по утверждённой структуре.

Используй только опорный материал ниже.
Не добавляй новые исследования, статистику,
кейсы, цитаты и причинные объяснения.

Если чего-то не хватает,
поставь [НУЖЕН ИСТОЧНИК]
или [НУЖЕН ПЕРЕХОД].

Опорный материал:
[вставить]
```

# 9. Несколько редакторов вместо команды «улучши»

## Структурный редактор

```text
Не переписывай текст.

Проверь:
- понятна ли главная мысль;
- выполняет ли каждый раздел отдельную функцию;
- логичен ли порядок;
- нет ли преждевременного вывода;
- где читатель потеряет нить;
- что можно объединить или удалить.

Верни:
фрагмент → проблема → структурное действие.
```

## Редактор аргументации

```text
Для каждого существенного вывода укажи:

- на каких основаниях он построен;
- достаточны ли основания;
- нет ли скачка от корреляции к причинности;
- рассмотрено ли сильнейшее возражение;
- не сильнее ли вывод, чем данные;
- что могло бы его опровергнуть.

Не оценивай стиль.
```

## Фактчек-редактор

```text
Извлеки внешние утверждения:
числа, даты, имена, цитаты,
результаты исследований, причинные связи и обобщения.

Для каждого укажи:
- нужен ли источник;
- есть ли он в тексте;
- что именно требуется проверить;
- можно ли ослабить формулировку,
  если подтверждение недостаточно.
```

## Читатель-скептик

```text
Назови три места,
где внимательный читатель потеряет доверие к тексту.

Для каждого объясни:
- что вызывает сомнение;
- какого основания не хватает;
- какая альтернативная интерпретация возможна;
- какое минимальное исправление поможет.
```

## Редактор тона

```text
Аудитория: [описать].
Цель текста: [описать].
Желаемый тон: уверенный, но не высокомерный;
живой, но не фамильярный;
критический, но не агрессивный.

Найди места, где тон нарушается.
Предложи минимальные замены.
Не переписывай текст целиком.
```

## Стилистический редактор

```text
Найди:
- канцелярит;
- повторы;
- пустые вводные фразы;
- слишком длинные предложения;
- необоснованные усилители;
- неясные местоимения;
- скачки регистра;
- штампы и нейросетево-гладкие формулировки.

Не меняй факты, позицию и структуру.
```

# 10. Итеративная редактура за пять проходов

## 1. Структура

```text
Найди ключевые мысли черновика.
Предложи порядок блоков,
но пока не переписывай предложения.

Покажи, что оставить, объединить,
переставить, удалить или добавить.
```

## 2. Логика

```text
Для каждого блока укажи:
тезис → основание → вывод → связь со следующим блоком.

Найди логические разрывы.
Не редактируй стиль.
```

## 3. Сокращение

```text
Удали повторы, пустые вводные,
лишние примеры и абзацы без новой мысли.

Не добавляй нового содержания.
После сокращения перечисли существенные потери.
```

## 4. Язык и тон

```text
Сделай текст яснее и естественнее.

Сохрани факты, позицию, структуру и голос автора.
Не превращай текст в рекламный,
слишком официальный или нейросетево-гладкий.
```

## 5. Сравнение версий

```text
Сравни исходную и итоговую версии.

Проверь:
- какие факты потерялись;
- где изменилась позиция;
- где вывод стал сильнее без основания;
- где текст стал точнее;
- где он стал только красивее.

Не считай последнюю версию лучшей автоматически.
```

# 11. Один материал — разные форматы

Адаптация — это не механическое сокращение. У каждого формата своя задача.

```text
У меня есть опорный материал.

Подготовь:
1. статью — подробно объяснить вопрос;
2. короткий пост — передать один главный вывод;
3. письмо коллеге — предложить действие;
4. пять слайдов — поддержать устное выступление;
5. памятку — дать последовательность действий.

Для каждого формата сначала укажи:
- аудиторию;
- цель;
- желаемое действие читателя;
- допустимый объём;
- что можно потерять;
- что терять нельзя.

Только после этого создавай тексты.
```

## Из длинного в короткое

```text
Адаптируй статью в сообщение для рабочего чата.

Сохрани:
- главный вывод;
- одно основание;
- одно ограничение;
- требуемое действие.

Удали историю вопроса,
второстепенные примеры и подробный обзор возражений.

Не добавляй разговорный тон и эмодзи автоматически.
```

## Из короткого в длинное

Короткий пост нельзя безопасно «раздуть» без дополнительных данных.

```text
Ниже короткий пост.

Не превращай его сразу в статью.

Сначала покажи:
- какие тезисы в нём есть;
- каких оснований не хватает;
- где нужен источник;
- какие примеры потребуются;
- какие разные статьи можно построить
  из этой формулировки.

После моего выбора подготовь структуру.
```

## Проверка потерь

```text
Сравни исходный материал и адаптацию.

Покажи:
- потерянные факты и условия;
- удалённые возражения;
- усиленные обещания;
- изменение аудитории;
- изменение степени уверенности;
- добавленные сведения.

Какие потери допустимы для нового формата,
а какие искажают смысл?
```

# 12. Финальный аудит

```text
Проведи финальный аудит текста.

СОДЕРЖАНИЕ
- отвечает ли текст на исходный вопрос;
- понятен ли главный вывод;
- нет ли противоречий.

ДОКАЗАТЕЛЬСТВА
- подтверждены ли числа и цитаты;
- не сильнее ли выводы, чем источники;
- отделены ли факты от интерпретаций.

РЕДАКТУРА
- нет ли повторов и пустых абзацев;
- соответствует ли тон аудитории;
- сохранился ли голос автора.

АДАПТАЦИЯ
- не потеряны ли важные ограничения;
- не появились ли новые обещания;
- понятно ли требуемое действие читателя.

Сначала перечисли только существенные проблемы.
Не переписывай текст без отдельной команды.
```

# Главное

ИИ полезен не как машина, которая мгновенно пишет финальную статью, а как набор разных рабочих ролей:

```text
исследователь
→ аналитик источников
→ синтезатор
→ архитектор текста
→ автор черновика
→ структурный редактор
→ критик аргументации
→ фактчекер
→ стилист
→ адаптер формата
```

Эти роли лучше включать последовательно. Если попросить всё сразу, модель смешает исследование, сочинение, проверку и редактуру — и станет трудно понять, где закончились источники и началась правдоподобная импровизация.
