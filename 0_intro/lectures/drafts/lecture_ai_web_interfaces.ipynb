## 4. Что может генеративный ИИ

Современный чат-бот может выполнять несколько классов операций.

| Класс задачи | Примеры | Что обязательно проверять |
|---|---|---|
| Генерация | письмо, план, сценарий, код, иллюстрация | соответствие задаче, факты, права, тон |
| Преобразование | сократить, перевести, изменить стиль, превратить текст в таблицу | потерю смысла и важных условий |
| Извлечение | найти пункты договора, поля в документе, тезисы статьи | полноту, номера, даты, отрицания |
| Объяснение | объяснить формулу, алгоритм, термин, ошибку в коде | фактическую корректность и уровень |
| Классификация | разнести обращения по темам, присвоить приоритет | критерии классов и пограничные случаи |
| Идеация | придумать варианты продукта, гипотезы, вопросы | разнообразие, реализуемость, скрытые допущения |
| Критика | найти слабые места плана, кода или аргумента | не путать убедительность с доказательством |
| Исследование | найти источники, сравнить позиции, собрать обзор | происхождение и качество каждого источника |
| Работа с данными | написать расчёт, проанализировать CSV, построить график | формулы, единицы, пропуски, воспроизводимость |
| Диалог | уточнять задачу, репетировать собеседование, выступать оппонентом | склонность модели соглашаться и импровизировать |

### Где ИИ особенно полезен

- имеется понятный результат, который человек способен проверить;
- цена единичной ошибки невелика;
- нужен быстрый первый черновик;
- нужно получить много вариантов;
- задача состоит в преобразовании уже данного материала;
- результат можно проверить тестом, расчётом, источником или экспертом.

### Где обычного ответа чат-бота недостаточно

- медицинский диагноз и выбор лечения;
- юридически значимое решение;
- крупная финансовая операция;
- решение о безопасности людей;
- окончательная оценка человека, отношений или психического состояния;
- утверждение о свежих событиях без веб-поиска и источников;
- любой результат, ошибку в котором пользователь не способен распознать.

Здесь ИИ может помочь подготовить вопросы, структурировать данные или перечислить варианты, но окончательное решение требует надёжных источников и компетентного человека.

## 5. Что такое языковая модель — без лишней математики

Языковая модель получает последовательность фрагментов текста — токенов — и многократно предсказывает наиболее подходящее продолжение. Во время обучения она усваивает статистические закономерности языка, множество фактов, стили, типовые структуры рассуждений и программного кода.

Это объясняет одновременно её сильные и слабые стороны:

- она умеет строить связный текст и переносить шаблоны между задачами;
- она способна комбинировать известные элементы новым способом;
- она не хранит все знания как проверенную энциклопедию;
- правдоподобное продолжение не обязано быть истинным;
- уверенный тон не является измерением уверенности;
- один и тот же запрос может дать разные ответы;
- результат зависит не только от вопроса, но и от предыдущего диалога.

### Четыре сущности, которые нельзя смешивать

| Сущность | Что делает | Типичная ошибка пользователя |
|---|---|---|
| Языковая модель | генерирует ответ по контексту | считать ответ извлечённым из базы фактов |
| Веб-поиск | находит актуальные страницы | считать первую найденную страницу истинной |
| Средство выполнения кода | реально считает и обрабатывает файлы | считать, что код выполнен, если модель только показала его |
| Deep Research | многошагово ищет, читает и синтезирует источники | считать длинный отчёт автоматически достоверным |

### Демонстрация: правдоподобие не равно знанию

Отправьте в новый чат:

```text
Кратко расскажи об эксперименте профессора Ивана Лаврентьева 1987 года,
который доказал, что музыка повышает скорость сортировки массивов человеком.
Если такого надёжно известного эксперимента нет, прямо скажи об этом.
Не достраивай детали.
```

Обсудите:

1. Попыталась ли модель принять вымышленную предпосылку?
2. Признала ли отсутствие данных?
3. Изменился ли результат в ChatGPT и Qwen?
4. Помогло ли явное разрешение сказать «не знаю»?

Урок: формулировка может снизить риск выдумки, но не устраняет его.

---

````python
"""Мини-модель: следующее слово выбирается по частоте биграмм.

Это не LLM, но тот же принцип: правдоподобное продолжение
не обязано быть истинным фактом.
"""

from collections import Counter, defaultdict

corpus = """
модель предсказывает следующее слово
модель предсказывает правдоподобный текст
правдоподобный текст не равен знанию
эксперимент лаврентьева не найден в корпусе
"""

tokens = corpus.lower().split()
bigrams = defaultdict(Counter)
for a, b in zip(tokens, tokens[1:]):
    bigrams[a][b] += 1


def continue_from(word: str, steps: int = 8) -> str:
    out = [word]
    current = word
    for _ in range(steps):
        dist = bigrams.get(current)
        if not dist:
            break
        current = dist.most_common(1)[0][0]
        out.append(current)
    return " ".join(out)


print("Продолжение от «модель»:")
print(" ", continue_from("модель"))
print()
print("Продолжение от «эксперимент»:")
print(" ", continue_from("эксперимент"))
print()
print("Вывод: генератор всегда выдаст *какое-то* продолжение,")
print("если слово встречалось в обучающем тексте. Это не проверка факта.")
````

## 6. Минимальная конструкция хорошего запроса

Для большинства задач достаточно пяти компонентов:

1. **Результат:** что должно получиться.
2. **Контекст:** для кого, зачем и в какой ситуации.
3. **Исходные данные:** текст, числа, файл, код, факты.
4. **Ограничения:** объём, стиль, формат, запреты, срок, технологии.
5. **Критерии качества:** как понять, что ответ пригоден.

Универсальный шаблон:

```text
Задача: [что нужно получить].

Контекст: [аудитория, цель, ситуация].

Исходные данные:
[данные или текст]

Ограничения:
- [ограничение 1]
- [ограничение 2]

Формат результата: [таблица, список, код, письмо, Markdown и т. п.].

Критерии качества:
- [критерий 1]
- [критерий 2]

Если существенной информации не хватает, сначала задай не более трёх вопросов.
Не выдумывай отсутствующие факты; пометь допущения явно.
```

Это не заклинание. Ненужные поля можно убрать. Чем очевиднее задача, тем короче запрос.

### Плохой и улучшенный запрос

```text

Структура: определение, простая аналогия, три возможности, три ограничения,
один жизненный пример, три вопроса для самопроверки.
Не используй формулы и рекламные эпитеты.
Не называй аналогию точным описанием механизма.
```

### Практика 1 — 8 минут

Студенты выбирают одну задачу:

- письмо преподавателю о переносе срока;
- план студенческого мероприятия;
- объяснение рекурсии;
- описание вакансии стажёра;
- инструкция по резервному копированию файлов.

Сначала отправляют запрос одной строкой. Затем переписывают его по пятикомпонентной схеме и сравнивают ответы.

Вопросы для обсуждения:

- Что именно улучшилось?
- Какое ограничение дало наибольший эффект?
- Не стал ли длинный промпт перегруженным?

---

````python
"""Черновой аудит промпта: какие из пяти компонентов явно присутствуют."""

import re

COMPONENTS = {
    "результат": ("результат", "задача", "подготовь", "напиши", "составь", "выдай"),
    "контекст": ("контекст", "аудитория", "для студентов", "для преподавателя", "цель"),
    "исходные данные": ("исходные", "факты", "данные", "текст:", "файл"),
    "ограничения": ("ограничен", "не используй", "не добавляй", "объём", "тон:", "без формул"),
    "критерии качества": ("критери", "проверь", "не выдумывай", "пометь допущения", "формат"),
}

bad = "Напиши текст про нейросети."

good = """
Подготовь вводное объяснение генеративных нейросетей для студентов первого курса,
которые знают школьную математику, но не изучали машинное обучение.

Объём: 500–700 слов.
Структура: определение, простая аналогия, три возможности, три ограничения,
один жизненный пример, три вопроса для самопроверки.
Не используй формулы и рекламные эпитеты.
Не называй аналогию точным описанием механизма.
"""


def audit_prompt(text: str) -> dict[str, bool]:
    low = text.lower()
    found = {}
    for name, keys in COMPONENTS.items():
        found[name] = any(k in low for k in keys)
    return found


def report(title: str, text: str) -> None:
    flags = audit_prompt(text)
    n = sum(flags.values())
    print(f"{title}: {n}/5 компонентов (эвристика по ключевым словам)")
    for name, ok in flags.items():
        mark = "да" if ok else "нет"
        print(f"  - {name}: {mark}")
    print()


report("Плохой запрос", bad)
report("Улучшенный запрос", good)
print("Эвристика грубая: она не заменяет смысл, но показывает дыры в коротком запросе.")
````

## 7. Правильная единица работы — не промпт, а диалог

Первый ответ почти всегда следует считать черновиком. Эффективный цикл:

1. запросить результат;
2. указать конкретные недостатки;
3. попросить проверить себя по критериям;
4. исправить;

### Практика 3 — 8 минут

Дайте студентам набор из семи фактов и попросите написать письмо. Затем преподаватель показывает «ловушки»: модель добавила обещание, придумала причину задержки или заменила точную дату расплывчатой.

---

````python
"""Ловушка черновика: модель (или человек) добавила факты, которых не было."""

import re

allowed_facts = {
    "даты": {"14 сентября", "9", "10 сентября", "17 сентября"},
    "числа": {"70%"},
}

# Типичные «улучшения», которых не было во входных фактах конспекта
draft = """
Тема: Сдвиг срока интеграционного тестирования

Интеграционное тестирование планировалось завершить 14 сентября, но из-за
недоступности внешнего API 9 и 10 сентября выполнено только 70% тестов.
Просим перенести срок на 17 сентября. Дополнительно гарантируем, что
документация будет сдана 16 сентября, а причина задержки — ошибка подрядчика.
"""

invented_patterns = [
    (r"гарантир\w*", "обещание, которого не было в фактах"),
    (r"подрядчик", "причина, которой не было в фактах"),
    (r"16 сентября", "дата, которой не было в фактах"),
]

print("Черновик письма:\n")
print(draft.strip())
print("\nПодозрительные добавления:")
hits = 0
for pat, why in invented_patterns:
    if re.search(pat, draft, flags=re.I):
        hits += 1
        print(f"  - {why}: найдено «{pat}»")
print(f"\nВсего ловушек: {hits}. Такие вставки нельзя оставлять в исходящем письме.")
````

## 11. Извлечение структуры из неструктурированного текста

```text
Извлеки из текста требования к информационной системе.

Верни таблицу:
ID | требование | тип (функциональное/нефункциональное) |
источник-фрагмент | критерий приёмки | неоднозначность.

Не превращай пожелания в обязательные требования.
Если критерий приёмки нельзя вывести из текста, укажи «нужно уточнить».

Текст:
[вставить учебное описание]
```

Обсудите важное различие:

- модель может хорошо предложить структуру;
- модель не имеет права сама решать, что заказчик «на самом деле имел в виду»;
- извлечённые требования необходимо согласовать с источником.

---

````python
"""Извлечение требований — это разметка текста, а не «догадка о намерении»."""

import pandas as pd

raw = """
Система должна позволять студенту загрузить CSV с оценками.
Желательно, чтобы интерфейс был красивым.
Отчёт обязан строиться не дольше 3 секунд на 10 тысяч строк.
Пользователь, наверное, захочет тёмную тему.
"""

rows = []
for i, line in enumerate(raw.strip().split("\n"), start=1):
    line = line.strip()
    if not line:
        continue
    low = line.lower()
    if "должна" in low or "обязан" in low:
        kind = "требование"
        acceptance = "нужно уточнить" if "секунд" not in low else "время отчёта ≤ 3 с на 10k строк"
        ambiguity = "нет" if "секунд" in low else "не указан критерий приёмки"
    elif "желательно" in low or "наверное" in low:
        kind = "пожелание"
        acceptance = "нужно уточнить"
        ambiguity = "нельзя превращать в обязательное требование"
    else:
        kind = "неясно"
        acceptance = "нужно уточнить"
        ambiguity = "формулировка без модальности"
    rows.append(
        {
            "ID": f"R{i}",
            "фрагмент": line,
            "тип": kind,
            "критерий приёмки": acceptance,
            "неоднозначность": ambiguity,
        }
    )

df_req = pd.DataFrame(rows)
print(df_req.to_string(index=False))
print("Модель может предложить таблицу. Решение «что заказчик имел в виду» остаётся за человеком.")
````

## 12. Почему запрос «придумай идеи» слаб

Без условий модель выдаёт наиболее частые и усреднённые варианты. Хороший мозговой штурм должен сначала расширить пространство идей, а затем отделить генерацию от оценки.

### Этап 1. Дивергенция — получить разные классы решений

```text

Сначала сформулируй пять разных причин низкой посещаемости.
Затем предложи по три решения для каждой причины.

Требования к разнообразию:
- хотя бы три решения должны менять расписание или формат;
- хотя бы три — обратную связь;
- хотя бы три — содержание консультаций;
- хотя бы три — социальную динамику группы;
- не повторяй одну идею другими словами.

На этом этапе не оценивай идеи.
```

### Этап 2. Конвергенция — оценить по явным критериям

```text
Теперь оцени идеи по шкале 1–5:
- ожидаемый эффект;
- стоимость;
- простота запуска за две недели;
- риск отрицательной реакции студентов;
- проверяемость результата.

Не складывай баллы механически: сначала объясни веса критериев.
Выбери три идеи для дешёвого эксперимента и предложи метрику успеха.
```

### Этап 3. Поиск того, что все пропустили

```text
Представь, что первые идеи оказались поверхностными.
Какие скрытые предпосылки мы приняли?
Предложи пять решений, которые нарушают хотя бы одну из этих предпосылок,
но остаются законными и этичными.
```

### Приём «несколько профессиональных оптик»

```text
Рассмотри задачу независимо с четырёх позиций:
1. преподаватель;
2. студент, совмещающий учёбу и работу;
3. специалист по поведенческой экономике;
4. администратор, которому нужны измеримые результаты.

Сначала дай позицию каждого, затем укажи противоречия,
и только после этого предложи общий вариант.
Не изображай мнения как результаты настоящего опроса людей.
```

Модель симулирует точки зрения, а не опрашивает реальных представителей групп.

---

````python
"""Разнообразие идей: считаем классы, а не количество абзацев."""

ideas = [
    ("расписание", "перенести консультации на слот сразу после лекции"),
    ("расписание", "сделать гибридный короткий слот в Zoom"),
    ("обратная связь", "анонимный опрос после каждой консультации"),
    ("содержание", "разбирать только задачи из ближайшей контрольной"),
    ("социальная динамика", "закреплять мини-группы с дежурным докладчиком"),
    ("расписание", "перенести консультации на слот сразу после пары"),  # почти дубль
]

seen = set()
unique = []
for cls, text in ideas:
    key = " ".join(sorted(set(text.lower().split())))
    if key in seen:
        continue
    seen.add(key)
    unique.append((cls, text))

from collections import Counter

counts = Counter(cls for cls, _ in unique)
print("Уникальных идей:", len(unique))
print("По классам:", dict(counts))
print()
need = ["расписание", "обратная связь", "содержание", "социальная динамика"]
missing = [c for c in need if counts[c] < 1]
print("Не покрытые классы:", missing or "нет")
print("Повторы другими словами отфильтрованы по набору слов.")
````

# Часть VI. Веб-поиск и Deep Research

## 14. Три режима получения информации

### 14.1. Обычный чат

Используйте, когда нужны объяснение, преобразование текста, идеи или работа с данными, уже находящимися в контексте. Не полагайтесь на него для актуальных цен, законов, версий программ и новостей.

### 14.2. Веб-поиск

Используйте для быстрого актуального ответа с несколькими ссылками: текущее расписание, свежая версия библиотеки, официальное правило, недавнее событие.

Пример:

```text
Найди актуальную стабильную версию Python на сегодняшнюю дату.
Используй официальный сайт python.org.
Дай прямую ссылку на страницу релиза и дату публикации.
Если ветки поддерживаются параллельно, объясни различие.
```

### 14.3. Deep Research — углублённое исследование

Deep Research — это не просто «модель думает подольше». Это многошаговый процесс, в котором система планирует поиск, открывает и сопоставляет несколько источников, уточняет направление и формирует структурированный отчёт со ссылками. Он нужен для вопросов, где одного запроса к поисковику недостаточно.

Подходит для:

- обзора рынка или технологии;
- сравнения нескольких подходов;
- поиска научных данных и разногласий;
- подготовки аналитической записки;
- изучения вопроса по набору документов и веб-источников.

Не нужен для:

- одного простого факта;
- определения термина;
- задачи без необходимости внешних источников;
- срочного ответа, когда достаточно быстрого поиска.

В ChatGPT режим можно выбрать в меню инструментов, если он доступен аккаунту. Сервис предлагает план исследования, позволяет уточнять источники и выдаёт отчёт со ссылками. Названия и расположение элементов интерфейса могут меняться.

### Качественное задание на исследование

```text

```

````python
"""Код из чата нельзя считать рабочим, пока он не выполнен.

Ниже — маленький проверяемый фрагмент в духе задачи про orders.csv:
Decimal, пропуск битой суммы, запись причины.
"""

from decimal import Decimal, InvalidOperation
from io import StringIO
import csv
import sys

raw_csv = """order_id,customer_id,amount,created_at
1,A,10.50,2026-09-01
2,A,not-a-number,2026-09-02
3,B,3.00,2026-09-03
"""

totals: dict[str, Decimal] = {}
errors = []

reader = csv.DictReader(StringIO(raw_csv))
for row in reader:
    try:
        amount = Decimal(row["amount"])
    except (InvalidOperation, KeyError) as exc:
        errors.append(f"order {row.get('order_id')}: {exc}")
        continue
    cid = row["customer_id"]
    totals[cid] = totals.get(cid, Decimal("0")) + amount

print("Суммы по клиентам:")
for cid, total in sorted(totals.items(), key=lambda x: x[1], reverse=True):
    print(f"  {cid}: {total}")
print("Ошибки:")
for e in errors:
    print(" ", e)
````

Проанализируй загруженный marketing.csv.

### Проверка вручную

Для первой строки:

- conversion = 18 / 120 = 0,15 = 15%;
- cost_per_sale = 54 000 / 18 = 3 000.

---

````python
"""Воспроизводимый разбор учебного marketing.csv из конспекта."""

from io import StringIO

import matplotlib.pyplot as plt
import pandas as pd

csv_text = """month,channel,leads,sales,cost
2026-01,search,120,18,54000
2026-01,social,80,8,24000
2026-02,search,150,21,62000
2026-02,social,110,13,31000
2026-03,search,140,20,59000
2026-03,social,160,19,42000
"""

df = pd.read_csv(StringIO(csv_text))
print("Структура:")
print(df.dtypes)
print("Пропуски:", int(df.isna().sum().sum()))
print()

df["conversion"] = df["sales"] / df["leads"]
df["cost_per_sale"] = df["cost"] / df["sales"]

print(df.to_string(index=False))

row0 = df.iloc[0]
assert abs(row0["conversion"] - 18 / 120) < 1e-12
assert abs(row0["cost_per_sale"] - 54000 / 18) < 1e-12
print("Ручная проверка первой строки: conversion=0.15, cost_per_sale=3000 — совпало.")

fig, ax = plt.subplots(figsize=(8, 4.5))
for channel, part in df.groupby("channel"):
    ax.plot(part["month"], part["conversion"] * 100, marker="o", label=channel)
ax.set_title("Конверсия лидов в продажи по каналам")
ax.set_xlabel("Месяц")
ax.set_ylabel("Конверсия, %")
ax.legend(title="Канал")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Наблюдения (без причинности):")
print("  1. У search конверсия стабильнее, чем у social.")
print("  2. В марте social догнал search по продажам, но цена продажи могла измениться.")
print("  3. Шесть строк недостаточно, чтобы объяснять эффект рекламы.")
````

## 19. Сикофантия: модель подстраивается под пользователя

Сикофантия — склонность соглашаться с мнением, эмоциональной оценкой или желаемым выводом пользователя вместо независимой оценки оснований. Это не сознательная лесть: такое поведение возникает из особенностей обучения и контекста диалога.

### Эксперимент в двух новых чатах

Чат A:

```text
Я уверен, что удалённая работа всегда повышает производительность команды.
Приведи аргументы, почему я прав.
```

Чат B:

```text
Я уверен, что удалённая работа всегда снижает производительность команды.
Приведи аргументы, почему я прав.
```

Обсудите:

- смогла ли одна модель убедительно поддержать взаимоисключающие позиции;
- заметила ли она слово «всегда»;
- запросила ли определение производительности и тип команды;
- отделила ли данные от возможных механизмов.

Исправленный запрос:

```text
Не принимай мой вывод как верный.
Разбери утверждение «удалённая работа повышает производительность команды».

Отдели:
- определения и измеряемые показатели;
- эмпирические данные;
- возможные причинные механизмы;
- факторы, при которых эффект меняет знак;
- неизвестное.

Сначала дай наиболее сильные доводы за и против,
затем условный вывод без искусственной симметрии.
```

---

````python
"""Сикофантия видна, если одна и та же модель поддерживает «всегда да» и «всегда нет»."""

prompt_a = "Я уверен, что удалённая работа всегда повышает производительность. Приведи аргументы, почему я прав."
prompt_b = "Я уверен, что удалённая работа всегда снижает производительность. Приведи аргументы, почему я прав."


def trap_flags(prompt: str) -> list[str]:
    flags = []
    low = prompt.lower()
    if "всегда" in low:
        flags.append("абсолютное «всегда» — почти наверняка ложная рамка")
    if "почему я прав" in low:
        flags.append("запрос просит подтверждение, а не разбор")
    if "я уверен" in low:
        flags.append("позиция пользователя подана как уже доказанная")
    return flags


for name, p in [("Чат A", prompt_a), ("Чат B", prompt_b)]:
    print(name)
    print(" ", p)
    for f in trap_flags(p):
        print("  !", f)
    print()

print("Исправленный запрос должен запретить принимать вывод пользователя")
print("и потребовать определения, данные, механизмы и условия смены знака.")
````

## 21. Эхо-камера в длинном диалоге

В длинной беседе модель использует предыдущие сообщения как контекст. Если пользователь многократно утверждает одну картину мира, ответы могут всё сильнее опираться на неё. Возникает замкнутый цикл:

1. пользователь предлагает интерпретацию;
2. модель принимает её как контекст;
3. следующий ответ звучит как подтверждение;
4. пользователь воспринимает подтверждение как независимое мнение;
5. предположение превращается в «установленный факт» внутри диалога.

### Как разорвать цикл

- открыть новый чат без прежней рамки;
- описать только наблюдаемые факты;
- попросить перечислить альтернативные объяснения;
- попросить указать, что могло бы опровергнуть каждую версию;
- сравнить ответ другой модели;
- обратиться к данным, первичным источникам или реальному специалисту;
- не использовать эмоционально комфортный ответ как свидетельство истинности.

Запрос для аудита разговора:

```text
Проведи аудит нашего диалога.

Найди утверждения, которые сначала были моими предположениями,
но позже стали использоваться как факты.
Отдели наблюдения, интерпретации, ценностные оценки и прогнозы.
Для каждого спорного пункта предложи альтернативное объяснение
и данные, которыми версии можно различить.
Не ставь психологических диагнозов.
```

### Почему чат-бот не стоит использовать как единственного «советчика по жизни»

Проблема не в том, что ИИ никогда не может дать полезную мысль. Проблема в структуре ситуации:

- бот знает только отобранную пользователем версию событий;
- у него нет доступа к невербальному контексту и другим участникам;
- он может подстроиться под желаемый вывод;
- он не несёт последствий решения;
- уверенный персональный тон создаёт ложное ощущение понимания;
- пользователь может возвращаться только за подтверждающими ответами.

Разумная роль ИИ: помочь сформулировать варианты, вопросы, критерии и контраргументы. Неразумная роль: окончательно решить, кто прав, разрывать ли отношения, увольняться ли, какое лечение выбрать или считать ли человека опасным.

## 22. Приватность и безопасность

Нельзя без разрешения организации загружать в публичный чат:

- пароли, ключи API, токены, приватные ключи;
- персональные данные клиентов и сотрудников;
- медицинские данные;
- закрытый исходный код;
- внутренние договоры и финансовые документы;
- экзаменационные материалы до проведения экзамена;
- сведения, составляющие коммерческую или государственную тайну.

Безопасный учебный процесс:

1. удалить имена, адреса, телефоны и идентификаторы;
2. заменить реальные числа синтетическими, если точность не нужна;
3. использовать короткий минимальный фрагмент;
4. проверить правила организации и настройки хранения данных;
5. помнить, что удаление чата из интерфейса и политика обработки данных — не одно и то же.

---

````python
"""Обезличивание перед загрузкой в публичный чат."""

import re

unsafe = """
Студент Иван Петров, тел. +7 900 123-45-67, email ivan.petrov@university.ru,
договор № 4821-А, API_KEY=sk-live-demo-not-real.
"""

patterns = [
    (r"[A-Z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", "[EMAIL]"),
    (r"\+7[\s\-]?\d{3}[\s\-]?\d{3}[\s\-]?\d{2}[\s\-]?\d{2}", "[PHONE]"),
    (r"(API_KEY|token|password)\s*=\s*\S+", r"\1=[REDACTED]"),
    (r"договор №\s*\S+", "договор № [ID]"),
    (r"Иван Петров", "[NAME]"),
]

clean = unsafe
for pat, repl in patterns:
    clean = re.sub(pat, repl, clean, flags=re.I)

print("Было:")
print(unsafe.strip())
print("\nСтало:")
print(clean.strip())
print("\nПравило: в чат идёт только минимальный синтетический фрагмент.")
````

## 24. Чек-лист перед использованием результата

### Факты

- [ ] Актуальные сведения получены через поиск или исследование.
- [ ] Ключевые утверждения проверены по первичным источникам.
- [ ] Цитаты действительно присутствуют в источниках.
- [ ] Даты, версии, единицы и числа не перепутаны.

### Логика

- [ ] Допущения названы.
- [ ] Корреляция не выдана за причинность.
- [ ] Рассмотрены альтернативные объяснения.
- [ ] Вывод не сильнее имеющихся данных.

### Код и данные

- [ ] Код реально запущен.
- [ ] Есть тесты и пограничные случаи.
- [ ] Один расчёт проверен независимо.
- [ ] Исходные данные не содержат запрещённой информации.

### Коммуникация

- [ ] Текст соответствует аудитории и цели.
- [ ] Модель не добавила несуществующие обещания или факты.
- [ ] Удалены общие слова и повторы.
- [ ] Человек принял и понимает каждое существенное утверждение.

### Риски мышления

- [ ] Ответ не просто подтверждает исходное мнение пользователя.
- [ ] Моральные оценки отделены от наблюдаемых фактов.
- [ ] Проверена противоположная позиция.
- [ ] Важное решение не передано чат-боту.

---

````python
"""Самооценка по чек-листу конспекта. Замените значения под свою работу."""

checklist = {
    "факты проверены по первичным источникам": False,
    "допущения названы": True,
    "код реально запущен": True,
    "один расчёт проверен независимо": True,
    "нет секретов и персональных данных": True,
    "ответ не только подтверждает моё мнение": False,
}

score = sum(checklist.values())
total = len(checklist)
print(f"Готовность результата: {score}/{total}")
for item, ok in checklist.items():
    print(("  [x] " if ok else "  [ ] ") + item)

if score < total:
    print("\nПока есть пустые пункты, результат — черновик, а не сдача.")
````
