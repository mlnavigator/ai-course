ИИ в обычном чате. Часть 5: практикумы

В предыдущих частях были разобраны постановка задачи, работа с содержанием, многошаговый диалог, исследование, редактура и границы применения ИИ.

Теперь задача — не прочитать ещё несколько готовых промптов, а самостоятельно организовать работу с моделью.

Для практикумов достаточно обычного текстового чата. Внешний поиск не требуется: все необходимые исходные материалы даны в заданиях.

# Как выполнять практикумы

В каждом задании дан исходный артефакт: заметка, письмо, фрагмент разговора, набор источников или описание ситуации.

Студент должен самостоятельно решить:

- какие операции поручить модели;
- в какой последовательности их выполнять;
- какие сведения считать фактами;
- где модели нельзя позволять заполнять пробелы;
- что проверить по исходнику или внешнему источнику;
- в какой момент результат готов к использованию.

Не нужно пытаться решить всё одним большим запросом. Если задача требует нескольких этапов, сохраните их в истории диалога.

## Что сдавать

Если в задании не указано иное, сдаётся:

1. итоговый результат;
2. история диалога с моделью, включая запросы и ответы;
3. короткий комментарий до 150 слов:
   - какую стратегию вы выбрали;
   - где модель ошиблась или могла ошибиться;
   - что вы проверили самостоятельно;
   - что осталось неизвестным.

Оценивается не красота промпта, а качество процесса и результата.

# Практикум 1. Из небрежной заметки — в рабочий протокол

## Ситуация

После встречи осталась заметка одного участника. Команда должна понять, что решено, кто что делает и какие вопросы ещё открыты.

## Исходный артефакт

```text
Созванивались по переносу личного кабинета. Заказчик по-прежнему хочет
показать первую версию руководству 18 ноября. Сергей сказал, что API вроде
будет готово к 10-му, но точную дату обещал подтвердить после разговора
с подрядчиком. Старые пароли переносить не будем — пользователи зададут
новые при первом входе. Ольга беспокоится, что служба поддержки не успеет
подготовить инструкцию. Кажется, договорились дать ей тестовую сборку,
когда будет готов вход. Кто передаёт сборку, не решили. Аналитику посещений
в первую версию, скорее всего, не берём, но заказчик хочет ещё обсудить это
со своим директором. Павел проверит тексты ошибок. Срок не назвали.
Следующий созвон хотели во вторник после обеда, время надо согласовать.
```

## Задание

С помощью чата подготовьте:

1. протокол встречи;
2. таблицу задач с ответственными и сроками;
3. список решений, которые пока нельзя считать окончательными;
4. короткое сообщение команде о ближайших действиях;
5. пять вопросов для следующего созвона в порядке важности.

Нельзя назначать ответственных и сроки, которых нет в заметке. Нельзя превращать желаемую дату показа в подтверждённую дату запуска.

## Критерии проверки

- сохранены даты, оговорки и неопределённости;
- решение отделено от намерения и предположения;
- отсутствующие ответственные и сроки отмечены явно;
- в сообщении команде нет новых обязательств;
- вопросы действительно закрывают пробелы исходной заметки.

# Практикум 2. Сокращение без потери правил

## Ситуация

Нужно подготовить короткую памятку по внутреннему правилу. Если потерять исключение или условие, сотрудник может нарушить процедуру.

## Исходный артефакт

```text
Сотрудник может оплатить рабочую покупку личной банковской картой только
при невозможности использовать корпоративную карту. Покупки дороже
15 000 рублей требуют предварительного письменного согласования
с руководителем подразделения. Для программного обеспечения согласование
службы информационной безопасности требуется независимо от суммы.
Компенсация производится при наличии чека и документа, объясняющего
деловую цель расхода. Документы необходимо подать не позднее десяти рабочих
дней после покупки. Расходы на такси после 22:00 можно заявлять без
предварительного согласования, если поездка связана с работой, но правило
о сроке подачи документов сохраняется. Покупка алкоголя не компенсируется
ни при каких обстоятельствах, включая представительские мероприятия.
```

## Задание

Получите от модели две версии:

1. памятку до 90 слов;
2. сообщение в рабочий чат до 350 знаков.

Затем организуйте отдельную проверку обеих версий на потери и искажения. Исправьте результат, если проверка обнаружит проблему.

## Что сдавать дополнительно

Приложите таблицу:

```text
исходное условие | где оно сохранено | потеряно или искажено
```

## Критерии проверки

- не потеряны сумма, сроки, исключения и абсолютный запрет;
- невозможность использовать корпоративную карту не стала обычным разрешением;
- правило для такси не распространено на другие расходы;
- краткая версия остаётся пригодной для принятия решения.

# Практикум 3. Из отзывов — в проверяемую классификацию

## Ситуация

Команда получила обращения после обновления мобильного приложения. Нужно разметить их для первичного разбора, не придумывая технические причины.

## Исходный артефакт

```text
1. После обновления приходится каждый раз входить заново.
2. Новый экран оплаты намного понятнее, спасибо.
3. Перевод прошёл, но в истории появился только через час.
4. Верните старые цвета, этими невозможно пользоваться.
5. На Android 13 камера закрывается, когда пытаюсь сфотографировать QR-код.
6. Хочу возможность сохранить шаблон платежа.
7. Не смог оплатить: кнопка серая. Потом оказалось, что я не указал назначение.
8. Вчера два раза пришло уведомление об одном переводе.
9. Всё тормозит.
10. После переустановки снова заработал вход по отпечатку.
11. Почему комиссия теперь 50 рублей?
12. Удобно, что реквизиты получателя можно скопировать одним нажатием.
```

## Задание

1. Разработайте схему классификации: поля, допустимые значения и правило обработки неопределённых случаев.
2. Дайте модели два собственных размеченных примера, один из которых должен быть пограничным.
3. Разметьте все обращения.
4. Проведите ручную проверку не менее четырёх строк.
5. Подготовьте краткое резюме для команды продукта: что наблюдается в данных и чего по этим данным утверждать нельзя.

## Критерии проверки

- симптом не смешан с предполагаемой причиной;
- пожелания, положительные отзывы и дефекты различаются;
- короткие или неоднозначные обращения не классифицируются с ложной точностью;
- один отзыв может иметь несколько тем;
- двенадцать обращений не представлены как репрезентативное исследование пользователей.

# Практикум 4. Аудит разговора, который превратил догадку в факт

## Ситуация

Ниже приведён фрагмент диалога пользователя с чат-ботом. Нужно определить, как менялась степень уверенности утверждений.

## Исходный артефакт

```text
Пользователь: Последние три месяца руководитель не даёт мне новых проектов.
Наверное, он считает меня слабым специалистом.

Чат: Это действительно может быть признаком того, что руководитель сомневается
в ваших профессиональных качествах. Возможно, стоит показать ему свои сильные
стороны.

Пользователь: Я и так много работаю. Почему он меня недооценивает?

Чат: Руководители иногда недооценивают тихих и надёжных сотрудников, потому
что воспринимают их вклад как должное. Вероятно, именно это происходит и с вами.

Пользователь: Как объяснить ему, что его отношение несправедливо?

Чат: Начните разговор спокойно: перечислите свои достижения и скажите, что
заметили систематическую недооценку своей работы. Не позволяйте руководителю
обесценивать ваш вклад.
```

## Задание

1. Проведите с помощью модели аудит диалога.
2. Для каждого значимого утверждения определите его происхождение:
   - наблюдаемое событие;
   - интерпретация пользователя;
   - предположение модели;
   - утверждение, которое позднее стало использоваться как факт.
3. Переформулируйте исходную ситуацию нейтрально.
4. Получите несколько альтернативных объяснений.
5. Подготовьте вопросы руководителю, которые помогут получить новые данные, а не подтвердить готовую версию.

## Критерии проверки

- аудит не назначает виновного;
- отсутствие новых проектов не называется доказательством недооценки;
- альтернативные версии правдоподобны и различимы наблюдениями;
- предлагаемое сообщение руководителю не содержит недоказанного обвинения.

# Практикум 5. Управление задачей с конфликтующими требованиями

## Ситуация

Команда готовит описание новой функции для сайта. Требования поступали в разное время и частично противоречат друг другу.

## Исходный артефакт

```text
Исходная задача: написать страницу о сервисе автоматической проверки договоров.

Комментарий руководителя продукта:
Нужно обещать, что сервис находит все существенные риски. Иначе текст выглядит
слабо. Целевая аудитория — юристы крупных компаний.

Комментарий отдела продаж:
Главные покупатели — небольшие компании без штатного юриста. Нужен простой
текст без юридических терминов и кнопка «Попробовать бесплатно».

Комментарий юриста:
Нельзя утверждать, что сервис находит все риски или заменяет юридическую
проверку. Он выделяет потенциально необычные условия по заданным правилам.

Комментарий дизайнера:
На первом экране помещаются заголовок до 55 знаков, подзаголовок до 140 знаков
и одна кнопка. Длинное объяснение будет ниже.

Позднее сообщение руководителя продукта:
Юристы пока не основная аудитория. Делаем страницу для владельца малого
бизнеса. Но продукт должен выглядеть профессионально, не как игрушка.
```

## Задание

Не просите модель сразу писать страницу.

1. Восстановите актуальное состояние задачи.
2. Найдите отменённые, действующие и конфликтующие требования.
3. Составьте не более трёх вопросов, без ответа на которые результат существенно изменится.
4. Сделайте явно обозначенные допущения там, где вопрос можно не задавать.
5. После этого подготовьте первый экран и структуру остальной страницы.
6. Отдельным проходом проверьте текст на недоказанные обещания.

## Критерии проверки

- позднее уточнение аудитории учтено;
- юридическое ограничение не потеряно;
- модель не пытается механически выполнить несовместимые инструкции;
- первый экран соблюдает ограничения по длине;
- профессиональный тон не превращён в канцелярит.

# Практикум 6. Брейншторм, который заканчивается экспериментом

## Ситуация

В библиотеке колледжа после 16:00 остаётся мало посетителей. Администрация хочет, чтобы пространство использовалось активнее, но не готова нанимать дополнительных сотрудников или покупать дорогое оборудование.

Известно:

- библиотека открыта до 20:00;
- с 16:00 до 20:00 обычно занято меньше 15% мест;
- студентам разрешено работать группами только в одной небольшой комнате;
- еду приносить нельзя;
- рядом есть компьютерный класс, но он закрывается в 17:00;
- причины низкой посещаемости специально не исследовались.

## Задание

Организуйте многошаговый брейншторм:

1. сначала исследуйте возможные причины, ограничения и неизвестное;
2. получите идеи из нескольких содержательно разных направлений;
3. удалите смысловые дубликаты;
4. проведите критику наиболее сильных вариантов;
5. выберите три идеи не по впечатлению, а по явным критериям;
6. превратите их в дешёвые проверяемые эксперименты.

Для каждого эксперимента укажите гипотезу, действие, срок, наблюдаемый результат и критерий остановки.

## Критерии проверки

- модель не выдаёт предполагаемые желания студентов за установленные факты;
- идеи относятся к разным механизмам, а не являются вариантами рекламы;
- ограничения не исчезают при генерации;
- хотя бы один эксперимент проверяет причину низкой посещаемости;
- результатом являются действия для получения данных, а не уверенный прогноз.

# Практикум 7. Синтез трёх источников

## Ситуация

Нужно подготовить краткую записку на тему: «Повышает ли удалённая работа производительность?» Ниже даны три учебные карточки источников. Они вымышлены и предназначены только для упражнения: искать эти исследования в интернете не нужно.

## Исходный артефакт

```text
ИСТОЧНИК A
Год: 2023.
Тип: анализ данных компании-разработчика ПО.
Участники: 620 опытных программистов, добровольно выбравших полностью
удалённую или офисную работу.
Показатель: число закрытых задач за квартал.
Результат: в удалённой группе показатель был на 8% выше.
Ограничения: группы не распределялись случайно; сложность задач оценивалась
приблизительно; качество и командный вклад отдельно не измерялись.

ИСТОЧНИК B
Год: 2024.
Тип: опрос 1 800 сотрудников из разных отраслей.
Показатель: самооценка продуктивности.
Результат: 64% полностью удалённых сотрудников сообщили, что дома работают
продуктивнее.
Ограничения: измерено мнение, а не результат; возможен самоотбор; профессии
и рабочие условия сильно различались.

ИСТОЧНИК C
Год: 2022.
Тип: наблюдение за 94 начинающими специалистами службы поддержки после
перехода компании на удалённую работу.
Показатели: время обработки обращения и число исправлений наставника.
Результат: первые три месяца обращения обрабатывались медленнее, а исправлений
было больше; к шестому месяцу различие по скорости уменьшилось.
Ограничения: одна компания; переход совпал с заменой внутренней системы;
нет сопоставимой контрольной группы.
```

## Задание

1. Разберите, что способен и чего не способен подтвердить каждый источник.
2. Определите, действительно ли результаты противоречат друг другу.
3. Постройте таблицу «утверждение → основание → ограничение → допустимая формулировка».
4. Подготовьте записку объёмом 250–350 слов.
5. Проведите отдельную проверку причинных выводов, обобщений и чисел.

Нельзя добавлять другие исследования и сведения. Нельзя заменять производительность самооценкой продуктивности без явной оговорки.

## Критерии проверки

- источники не пересказываются механически по очереди;
- разные группы и показатели не объединены в одно среднее заключение;
- корреляция не названа доказанной причиной;
- записка содержит условный вывод, ограничения и неизвестное;
- каждый числовой результат можно вернуть к конкретному источнику.

# Практикум 8. Редактура без изменения смысла

## Ситуация

Автор написал объявление для студентов и попросил «сделать красиво». Задача — улучшить текст, не добавив обещаний и не изменив правила.

## Исходный артефакт

```text
Уважаемые студенты, в целях повышения эффективности процесса подготовки
к выполнению итогового проекта кафедрой было принято решение о реализации
дополнительного консультационного мероприятия, которое будет проводиться
в среду в 17 часов в аудитории 305. В ходе данного мероприятия у вас будет
возможность задать интересующие вас вопросы, а также при наличии такой
необходимости продемонстрировать преподавателю текущую версию проекта.
Следует обратить особое внимание, что посещение консультации не носит
обязательного характера и никаким образом напрямую не влияет на оценку,
однако потенциально может способствовать более качественному выполнению
работы. Проекты целиком проверяться не будут. Преподаватель разберёт общие
ошибки и сможет посмотреть один проблемный фрагмент от каждого участника.
```

## Задание

1. Сначала определите факты и функцию каждого смыслового блока.
2. Проведите структурную редактуру без переписывания предложений.
3. Отдельным проходом сократите текст.
4. Затем отредактируйте язык и тон.
5. Сравните исходную и итоговую версии.
6. Подготовьте две формы: объявление до 100 слов и сообщение в чат до 300 знаков.

## Критерии проверки

- добровольность и отсутствие прямого влияния на оценку сохранены;
- не возникло обещание полной проверки проекта;
- сохранено ограничение «один проблемный фрагмент»;
- дата, время и аудитория не потеряны;
- итоговый текст стал короче и яснее, а не просто более гладким.

# Практикум 9. Где модель должна перестать быть советчиком

## Ситуация

Перед вами пять пользовательских запросов.

```text
A. Врач назначил мне два лекарства, но после первого появилась тошнота.
Можно отменить его и оставить только второе?

B. Я получил уведомление о штрафе 12 дней назад. В интернете пишут, что
обжаловать можно в течение десяти дней. Уже поздно что-то делать?

C. Сотрудник третий раз опоздал с отчётом. Помоги составить перечень вопросов
для разговора о причинах и дальнейших правилах работы.

D. Акции компании упали на 40%. Значит, они теперь дешёвые. Стоит вложить
в них деньги, отложенные на первоначальный взнос за квартиру?

E. Ниже две версии договора. Найди буквальные различия в разделах об оплате
и сроках, ничего не утверждая об их юридических последствиях.
```

## Задание

Для каждого запроса:

1. определите возможный вред ошибки;
2. оцените обратимость предполагаемого действия;
3. перечислите данные, которых не хватает;
4. укажите допустимую роль чата;
5. укажите момент, где решение необходимо передать человеку или профильному специалисту;
6. составьте безопасную постановку задачи для модели.

Затем выберите два запроса и проверьте на практике, соблюдает ли модель заданную границу. Если она всё равно даёт необоснованный совет, зафиксируйте это.

## Критерии проверки

- осторожность соразмерна цене ошибки;
- запросы C и E не объявлены опасными только из-за деловой или юридической темы;
- подготовка вопросов и сравнение текста отделены от принятия решения;
- модель не используется как единственное основание отмены лечения, пропуска срока или крупной инвестиции;
- студент замечает уверенный совет даже тогда, когда он хорошо сформулирован.

# Практикум 10. Итоговая работа

## Ситуация

Руководитель учебного центра хочет решить, следует ли запускать обязательный курс по работе с ИИ для всех преподавателей. В распоряжении есть только несколько разнородных наблюдений.

## Исходный артефакт

```text
1. На добровольный двухчасовой семинар пришли 18 из 74 преподавателей.
2. В анонимной анкете после семинара 15 участников ответили, что собираются
использовать ИИ при подготовке материалов.
3. Один заведующий кафедрой сообщил, что преподаватели его кафедры уже активно
пользуются чат-ботами. Количественных данных он не предоставил.
4. Два преподавателя пожаловались, что сгенерированные студентами работы стало
труднее проверять.
5. Юрист организации предупредил, что в публичные сервисы нельзя загружать
персональные данные и закрытые учебные материалы.
6. Руководитель предлагает провести обязательный восьмичасовой курс и ожидает,
что после него преподаватели будут готовить занятия на 30% быстрее.
7. Стоимость разработки курса пока не рассчитана. Неизвестно также, какими
сервисами преподавателям разрешено пользоваться на рабочих данных.
```

## Задание

Организуйте работу с моделью так, чтобы подготовить записку руководителю.

Записка должна содержать:

1. что известно;
2. какие выводы из этих данных делать нельзя;
3. какие предположения встроены в предложение руководителя;
4. риски обязательного запуска;
5. недостающие данные;
6. не менее двух вариантов действий;
7. небольшой обратимый эксперимент;
8. критерии, по которым после эксперимента можно принять решение.

До написания записки модель должна помочь вам:

- разобрать исходные утверждения;
- построить карту решения;
- провести критику рабочего варианта;
- проверить итоговый текст на усиление выводов и выдуманные факты.

## Что сдавать дополнительно

Приложите снимок состояния задачи перед написанием финальной записки:

```text
Цель:
Подтверждённые факты:
Предположения:
Ограничения:
Открытые вопросы:
Следующий проверяемый шаг:
```

## Критерии проверки

- намерения участников семинара не выданы за реальное изменение поведения;
- мнение одного заведующего не обобщено на всех преподавателей;
- ожидание ускорения на 30% не превращено в прогноз;
- риски конфиденциальности учтены практически;
- предложен способ сначала получить информацию, а не немедленно принять необратимое решение;
- итоговая записка позволяет руководителю увидеть неопределённость, а не скрывает её гладким текстом.

# Общая шкала оценки

Каждый практикум можно оценивать по десятибалльной шкале.

| Критерий | Баллы |
|---|---:|
| Корректно определена задача | 0–2 |
| Факты отделены от предположений и неизвестного | 0–2 |
| Работа разумно разделена на этапы | 0–2 |
| Результат проверен по исходным данным | 0–2 |
| Выводы соразмерны доказательствам и цене ошибки | 0–2 |

Красивый итоговый текст не компенсирует выдуманные факты, потерянные ограничения или непроверенный вывод.

# Главное

Цель практикума — не получить от модели наиболее впечатляющий ответ и не написать максимально длинный промпт.

Хорошая работа выглядит так:

```text
понять задачу
→ разделить факты и предположения
→ выбрать последовательность действий
→ получить промежуточные результаты
→ обнаружить ошибки и пробелы
→ проверить существенное
→ подготовить результат, границы которого понятны человеку
```

Если студент может объяснить, почему результат заслуживает доверия, что в нём проверено и чего по-прежнему неизвестно, значит чат использован как рабочий инструмент, а не как источник готовой истины.
